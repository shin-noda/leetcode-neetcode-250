```python
class Solution:
    def reorganizeString(self, s):
        # Idea:
        # The bottleneck is the letter with the highest frequency

        n = len(s)
        letter_map = {}

        for i in range(n):
            letter_map[s[i]] = letter_map.get(s[i], 0) + 1

        occurrences = []
        for key in letter_map:
            occurrences.append((letter_map[key], key))

        # Sort by the occurrences (first element): O(26) = O(1)
        occurrences.sort(key=lambda x: x[0], reverse=True)

        max_oc = occurrences[0][0]
        counts = n - max_oc

        # We need at least (max_oc - 1) letters to meet the conditions.
        # For example, let's say 'a' has the highest occurences in s and
        # there are four 'a's. To meet the condition, we place 'a's like this:
        # a letters a letters a letters a
        #
        # To meet the condition, we need at least (4 - 1) letters in s
        # to meet the condition to use the letters as a partition.
        if not (counts >= max_oc - 1):
            # We don't have enough remaining letters in s to partion the
            # highest frequent letter.
            return ""

        # Create a list word all letters that consists of s.
        letters = []

        # The first is O(26), so O(26 N) = O(N) for a nested loop
        for i in range(len(occurrences)):
            oc, letter = occurrences[i]

            for _ in range(oc):
                letters.append(letter)

        partitions = [[] for _ in range(max_oc)]

        for i in range(len(letters)):
            partitions[i % max_oc].append(letters[i])

        # Finally, we combine all letters in the partitions
        results = []
        for blocks in partitions:
            # Note: we need to use .extend() instead of .append()
            # so that we don't append this list in the list.
            results.extend(blocks)

        return "".join(results)
```
