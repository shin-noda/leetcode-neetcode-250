```python
class TimeMap:

    def __init__(self):
        # Using a list of tuples (timestamp, value) to keep things ordered
        self.store = {}


    def set(self, key, value, timestamp):
        if key not in self.store:
            self.store[key] = []

        # Since timestamps are strictly increasing, we just append
        self.store[key].append((timestamp, value))


    def get(self, key, timestamp):
        if key not in self.store:
            return ""

        values = self.store[key]
        left, right = 0, len(values) - 1
        val = ""

        # Binary Search for the floor value
        while left <= right:
            mid = (left + right) // 2

            if values[mid][0] <= timestamp:
                # This is a potential candiate, look for a closer(larger) one
                val = values[mid][1]
                left = mid + 1

            else:
                right = mid - 1

        return val
```
