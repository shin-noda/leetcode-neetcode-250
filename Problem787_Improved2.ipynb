```python
import heapq

class Solution:
    def findCheapestPrice(self, n, flights, src, dst, k):
        # Build adjency list
        adj = [[] for _ in range(n)]
        for u, v, w in flights:
            adj[u].append((v, w))

        # Track the minimum stops taken to reach a node so far
        # Initialize with infinity
        visited_stops = [float('inf')] * n

        # Min-heap stores: (total_cost, current_node, stops_taken)
        min_heap = [(0, src, 0)]

        while min_heap:
            cost, curr, stops = heapq.heappop(min_heap)

            # Optimization: Because it's a min_heap, the first time we pop
            # the diestination, it is guaranteed to be the cheapest valid path
            if curr == dst:
                return cost

            # If we've hit the stop limit, we can't fly out of this city
            if stops > k:
                continue

            # Only explore neighbours if this path reached curr
            # using fewer stops than any previous path
            if stops >= visited_stops[curr]:
                continue

            visited_stops[curr] = stops

            # Explore neighbors
            for next_node, price in adj[curr]:
                heapq.heappush(min_heap, (cost + price, next_node, stops + 1))

        return -1
```
