```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None
```

```python
from collections import deque

class Codec:
    def serialize(self, root):
        """Encodes a tree to a single string.

        :type root: TreeNode
        :rtype: str
        """

        # Check if root is empty
        if not root:
            return ""

        results = []
        dq = deque([root])
        while dq:
            n = len(dq)

            for _ in range(n):
                node = dq.popleft()

                if node:
                    results.append(str(node.val))

                    dq.append(node.left)
                    dq.append(node.right)

                else:
                    results.append(str(node))

        # We trim unnecessarly "None"s at the end of the results
        while results:
            if results[-1] == "None":
                results.pop()

            else:
                break

        return ",".join(results)


    def deserialize(self, data):
        """Decodes your encoded data to tree.

        :type data: str
        :rtype: TreeNode
        """

        if not data:
            return None

        nodes = data.split(',')
        n = len(nodes)

        root = TreeNode(nodes[0])
        dq = deque([root])
        pointer = 1

        while dq and pointer < n:
            size = len(dq)

            for _ in range(size):
                # A potential parent node (could be a leaf)
                node = dq.popleft()

                if pointer < n:
                    # Left node
                    left_val = nodes[pointer]
                    pointer += 1

                    if left_val != "None":
                        left = TreeNode(int(left_val))
                        node.left = left
                        dq.append(left)

                # Right node
                if pointer < n:
                    # Right node
                    right_val = nodes[pointer]
                    pointer += 1

                    if right_val != "None":
                        right = TreeNode(int(right_val))
                        node.right = right
                        dq.append(right)

        return root
```
