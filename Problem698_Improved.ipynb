```python
class Solution:
    def canPartitionKSubsets(self, nums, k):
        total_sum = sum(nums)
        if total_sum % k != 0:
            return False

        target = total_sum // k

        # Essential for pruning
        nums.sort(reverse=True)

        if nums[0] > target:
            return False

        n = len(nums)
        visited = [False] * n

        # k_left: how many buckets still need filling
        # current_sum: progress on the active bucket
        # start_index: where to start looking in nums (prevents checking backwards)
        def backtrack(k_left, current_sum, start_index):
            # Base Case: successfully filled k-1 buckets (the last one must be valid)
            if k_left == 1:
                return True

            # If current bucket is full, move to the next one
            if current_sum == target:
                # Reset sum and start index
                return backtrack(k_left - 1, 0, 0)

            for i in range(start_index, n):
                # Skip if already used or if it overfills the current bucket
                if visited[i] or current_sum + nums[i] > target:
                    continue

                # Pruning: Skip duplicate numbers that just failed in this position
                if i > start_index and nums[i] == nums[i - 1] and not visited[i - 1]:
                    continue

                # Choose
                visited[i] = True

                # Explore (we pass i + 1 so we only look forward)
                if backtrack(k_left, current_sum + nums[i], i + 1):
                    return True

                # Backtrack
                visited[i] = False

                # Pruning: If an empty bucket fails to take the first available element,
                # this entire branch is a dead end
                if current_sum == 0:
                    return False

            return False

        return backtrack(k, 0, 0)
```
