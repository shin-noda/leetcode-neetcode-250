```python
class Solution:
    def search(self, nums, target):
        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = left + (right - left) // 2

            if nums[mid] == target:
                return mid

            elif nums[mid] < target:
                left = mid + 1

            else:
                right = mid - 1

        return -1
```

```python
solution = Solution()
```

```python
nums = [-1,0,3,5,9,12]
target = 9

print(solution.search(nums, target))
```
