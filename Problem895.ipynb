```python
class FreqStack:

    def __init__(self):
        # A standard dictionary to track the current frequency of each value
        self.freq = {}

        # A dictionary where keys are frequencies (1, 2, 3,...) and
        # values are Stacks
        self.group = {}
        self.max_freq = 0


    def push(self, val):
        # Update the frequency of the value
        f = self.freq.get(val, 0) + 1
        self.freq[val] = f

        # Update max_freq if this is a new high
        if f > self.max_freq:
            self.max_freq = f

        # Add the value to the stack corresponding to its current frequency
        if f not in self.group:
            self.group[f] = []

        self.group[f].append(val)


    def pop(self):
        # Get the top element from the highest frequency stack
        val = self.group[self.max_freq].pop()

        # Decrement the frequency of that value in our counter
        self.freq[val] -= 1

        # If the highest frequency stack is now empty, drop the max_freq level
        if not self.group[self.max_freq]:
            self.max_freq -= 1

        return val
```
