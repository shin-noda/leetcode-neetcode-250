```python
class Solution:
    def longestDiverseString(self, a, b, c):
        abc = [(a, 'a'), (b, 'b'), (c, 'c')]

        # Sort by increasing (ascending) order by the occurrences
        abc.sort(key=lambda x: x[0])

        sum_val = a + b + c
        max_val = abc[-1][0]

        # Calculate the partitions you need to maximise the string length
        partitions_required = max_val // 2

        # Check if we need to discard the max frequent letters
        # Note: this is a potential max number of partitions
        partitions = sum_val - max_val

        if partitions < partitions_required:
            # We add 1 for one extra bucket
            partitions += 1

            # We need to trim the max frequent letters
            max_val = partitions * 2

        else:
            # We add 1 because this partitions is used to create buckets
            # which is (partions + 1)
            partitions = partitions_required + 1

        buckets = [[] for _ in range(partitions)]

        # All letters except the max character
        fillings = []

        # Append the 2nd frequent letters
        for _ in range(abc[-2][0]):
            fillings.append(abc[-2][1])

        # Append the 3rd frequent letters
        for _ in range(abc[-3][0]):
            fillings.append(abc[-3][1])

        # There are multiple possibilitis to fill slots
        # 1. Fill slots with only one character
        # A A slots A A slots A A slots -> A A B A A B A A C A A ...
        #
        # 2. Fill slots will 2 letters like
        # (second most frequenst letter can fill slots)
        # A A slots A A slots A A slots -> A A B B C C A A B C C C A A ...

        # New Idea:
        # Actually, we can just combine the 2nd and 3rd letters and put into
        # each backet one by by starting from the 2nd frequent letters.

        max_oc = [abc[-1][1] for _ in range(max_val)]

        # First, create partitions with the most frequent letters
        for i in range(partitions):
            if i == partitions - 1:
                # Pop until there is no more in max_oc
                while max_oc:
                    buckets[i].append(max_oc.pop())

            else:
                # Pop twice
                buckets[i].append(max_oc.pop())
                buckets[i].append(max_oc.pop())

        # Next, we fill slots with remaining letters one by one
        # starting with the 2nd frequent letter and then, the 3rd ones.
        for i in range(len(fillings)):
            buckets[i % partitions].append(fillings[i])

        # Combine all letters in buckets
        results = []
        for i in range(partitions):
            results.extend(buckets[i])

        return "".join(results)
```
