```python
class Solution:
    def stoneGameII(self, piles):
        # i: how far into the array we have already consumed.
        # M: how many piles you're allowed to take
        # At any state (i, M):
        # - The current player chooses 1 <= X <= 2M
        # - They immediately gain the sum of the first X remaining piles
        # - Then the game continues from (i + X, max(M, X))
        # - And now it becomes the opponent's turn

        # dp(i, M): best score difference current player can achieve
        # (from piles starting at i, with parameter M)

        n = len(piles)

        # We constantly need: sum(piles[i:i+X]) and if we
        # compute that repeatedly, it's slow.
        # So we precompute suffix sums.
        suffix_sums = [0] * (n + 1)
        for i in range(n - 1, -1, -1):
            suffix_sums[i] = suffix_sums[i + 1] + piles[i]

        # Create our own manual cache dictionary
        memo = {}

        def dfs(i, M):
            # Check if we've already solved this exact subproblem
            if (i, M) in memo:
                return memo[(i, M)]

            # Base Case: the current player can just take all the remaining piles.
            if i + 2 * M >= n:
                return suffix_sums[i]

            min_next_player_score = float('inf')
            for X in range(1, 2 * M + 1):
                next_player_score = dfs(i + X, max(M, X))
                min_next_player_score = min(min_next_player_score, next_player_score)

            # Store the result in our cache before returning
            memo[(i, M)] = suffix_sums[i] - min_next_player_score

            return memo[(i, M)]

        return dfs(0, 1)
```

```python
solution = Solution()
```

```python
piles = [2,7,9,4,4]

print(solution.stoneGameII(piles))
```
