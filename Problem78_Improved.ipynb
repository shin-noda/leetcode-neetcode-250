```python
class Solution:
    def subsets(self, nums):
        results = []
        solutions = []

        def backtrack(i):
            # Base case: we've made a decision for every element
            if i == len(nums):
                # Take a snapshot of the current subset
                results.append(solutions[:])
                return

            # Choice 1: Include nums[i]
            solutions.append(nums[i])
            backtrack(i + 1)

            # Unchoose
            solutions.pop()

            # Choice 2: Exclude nums[i]
            backtrack(i + 1)

        backtrack(0)

        return results
```
