```python
import heapq

class Solution:
    def mostBooked(self, n, meetings):
        # Half-Closed Time Interval:
        # e.g. [start, end) = [1, 3)
        # The starting time 1 is inclusive but the ending time 3 is exclusive.

        # Crucial step: Sort meetings by start time
        meetings.sort()

        # This stores the number of meetings held
        rooms = [0] * n

        # Initialize heaps
        # free_rooms stores just the room IDs. Initially, all rooms are free.
        available = list(range(n))

        # (end_meeting_time, room_id)
        occupied = []

        for start, end in meetings:
            duration = end - start

            # First, free up any rooms in 'occupied'
            # that finish before or at 'start'
            while occupied and occupied[0][0] <= start:
                _, room_id = heapq.heappop(occupied)
                heapq.heappush(available, room_id)

            # Now check if we have a room ready to go
            if available:
                room_id = heapq.heappop(available)
                heapq.heappush(occupied, (end, room_id))

            else:
                # No rooms available. We must wait for the earliest room to finish
                earliest_end_time, room_id = heapq.heappop(occupied)

                # The meeting is delayed, so its new end time shifs forward by its duration
                new_end = earliest_end_time + duration
                heapq.heappush(occupied, (new_end, room_id))

            # Keep track of which room took the meeting
            rooms[room_id] += 1

        # Find the room with the maximum meetings (and lowest ID in case of a tie)
        max_meetings = -1
        result_room = -1

        for room_id in range(n):
            if rooms[room_id] > max_meetings:
                max_meetings = rooms[room_id]
                result_room = room_id

        return result_room
```

```python
solution = Solution()
```

```python
n = 2
meetings = [[0,10],[1,5],[2,7],[3,4]]

print(solution.mostBooked(n, meetings))
```
