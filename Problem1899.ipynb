```python
class Solution:
    def mergeTriplets(self, triplets, target):
        # Idea:
        # Since we use the max() operation, if one of the elements in a triplet
        # has a greater value than the target value in the index, we
        # cannot lower it.
        # Thus, the (a1_i, a2_i, a3_i) can be used only when
        # a1_i <= target1, a2_i <= target2, and a3_i <= target.
        # Also, one of (a1, a2, a3) should be a target value.
        # Other wise, it donesn't contribute to anything.

        t1, t2, t3 = target
        is_t1, is_t2, is_t3 = False, False, False

        # O(N * 3) = O(N)
        for a1, a2, a3 in triplets:
            # If at least one elements is the target value,
            # it contributes to create the target.
            if a1 <= t1 and a2 <= t2 and a3 <= t3:
                if a1 == t1:
                    is_t1 = True

                if a2 == t2:
                    is_t2 = True

                if a3 == t3:
                    is_t3 = True

                # We can create the combination
                if is_t1 and is_t2 and is_t3:
                    return True

        return False
```

```python
solution = Solution()
```

```python
triplets = [[2,5,3],[1,8,4],[1,7,5]]
target = [2,7,5]

print(solution.mergeTriplets(triplets, target))
```
