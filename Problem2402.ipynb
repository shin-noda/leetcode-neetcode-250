```python
import heapq

class Solution:
    def mostBooked(self, n, meetings):
        # Half-Closed Time Interval:
        # e.g. [start, end) = [1, 3)
        # The starting time 1 is inclusive but the ending time 3 is exclusive.

        # Important
        # We sort meetings with the first  and all values of start_i are unique,
        # so we don't have to worry about the conflits of the meeting with the
        # same starting time.
        # I consider starting time as meeting ids.
        meetings.sort()

        # Record number of meetings
        num_meetings = {}

        # (end_time, room_id)
        # Note: end_time is an absolute time.
        min_heap = []

        for i in range(n):
            num_meetings[i] = 0
            min_heap.append((0, i))

        # There are two possibilities to before starting a meeting:
        # 1. There is at least one empty room.
        # 2. All rooms are occupied, so we need to wait.
        #
        # For Case 1, we can take an empty room with the lowest room id.
        # For Case 2, we wait until one of the rooms will be empty.

        # We modify the approach.
        # Instead of looping through number of meetings, we loop through time.
        # However, the condition of ending the loop is based on the meetings

        # On a second thought, nope.

        curr = 0
        for meeting in meetings:
            start, end = meeting[0], meeting[1]
            duration = end - start

            curr = max(curr, start, min_heap[0][0])

            # This is O(n) but n <= 100, so this is O(100) = O(1)
            for i in range(n):
                if min_heap[i][0] != 0 and min_heap[i][0] <= curr:
                    et, rid = min_heap[i]
                    min_heap[i] = (0, rid)

            # Heapify: O(n) but n <= 100, so this is O(100) = O(1)
            heapq.heapify(min_heap)

            et, rid = heapq.heappop(min_heap)
            num_meetings[rid] += 1

            heapq.heappush(min_heap, (curr + duration, rid))

        # This part is for just fininding the room with the max number held
        room_num = -1
        max_meetings = -1
        for i in range(n):
            # This is '>', not '>=' for a tie-breaker.
            if num_meetings[i] > max_meetings:
                max_meetings = num_meetings[i]
                room_num = i

        return room_num
```
