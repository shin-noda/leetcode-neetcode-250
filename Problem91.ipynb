```python
class Solution:
    def numDecodings(self, s):
        # Idea:
        # At every position, you can arrive there from either:
        # 1. decoding one digit
        # 2. decoding two digits

        if not s or s[0] == '0':
            return 0

        n = len(s)

        # dp[i] will store the number of ways to decode s[0:i]
        # e.g. s = "226"
        # dp[0] = ways to decode ""
        # dp[1] = ways to decode "2"
        # dp[2] = ways to decode "22"
        # dp[3] = ways to decode "226"

        dp = [0] * (n + 1)

        # Base cases
        # Empty string has 1 valid way
        dp[0] = 1

        # First character (we already know it's not '0')
        dp[1] = 1

        for i in range(2, n + 1):
            # Check single digit (s[i - 1])
            # (Note: s[i - 1] corresponds to the character at dp[i])
            if s[i - 1] != '0':
                dp[i] += dp[i - 1]

            # Check two digits (s[i - 2:i])
            two_digit = int(s[i - 2:i])
            if 10 <= two_digit <= 26:
                dp[i] += dp[i - 2]

        return dp[-1]
```

```python
solution = Solution()
```

```python
s = "12"

print(solution.numDecodings(s))
```
