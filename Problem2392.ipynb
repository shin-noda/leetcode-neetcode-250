```python
from collections import deque, defaultdict

class Solution:
    def buildMatrix(self, k, rowConditions, colConditions):
        # Idea:
        # We are NOT directly placing numbers into a matrix
        # We are really being asked:
        # - Can you find an ordering of numbers that satisfies all row + column rules?
        # To solve this problem:
        # 1. Find valid row ordering
        # 2. Find valid column ordering
        # 3. Combine them into matrix coordinates

        # Use Kahn's Algorithm

        # Topological Sort:
        # A way to arrange things in an order that respects dependencies.
        def topo_sort(conditions):
            # 1. Build the adjacency list and indegree array
            graph = defaultdict(list)

            # Numbers start from 1
            indegree = [0] * (k + 1)

            for u, v in conditions:
                graph[u].append(v)
                indegree[v] += 1

            # 2. Queue all nodes with 0 incoming dependencies
            queue = deque()
            order = []

            for i in range(1, k + 1):
                # We care about nodes with indegree of 0 because
                # they don't have dependencies and nothing blocks them.
                # -> They can go first.
                if indegree[i] == 0:
                    queue.append(i)

            # 3. Process the queue
            while queue:
                node = queue.popleft()
                order.append(node)

                for neighbour in graph[node]:
                    indegree[neighbour] -= 1

                    if indegree[neighbour] == 0:
                        queue.append(neighbour)

            # If order doesn't include all elements from 1 to k, a cycle exists.
            if len(order) != k:
                return []

            return order


        # Get the independent row and column orderings
        row_order = topo_sort(rowConditions)
        col_order = topo_sort(colConditions)

        # If either has a cycle, it's impossible to buidl the matrix
        if not row_order or not col_order:
            return []

        # 4. Map each number to its correspoinding row and col index
        row_pos = {}
        col_pos = {}

        size = len(row_order)
        for i in range(size):
            row_num = row_order[i]
            col_num = col_order[i]

            row_pos[row_num] = i
            col_pos[col_num] = i

        # 5. Build the final k x k matrix
        matrix = [[0] * k for _ in range(k)]
        for num in range(1, k + 1):
            r = row_pos[num]
            c = col_pos[num]

            matrix[r][c] = num

        return matrix
```
