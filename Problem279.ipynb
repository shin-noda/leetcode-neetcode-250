```python
class Solution:
    def numSquares(self, n):
        # Idea:
        # This is a variation of the stair case proble.
        # If the level is a perfect square number, you can arrive the by 1 step.
        #

        dp = [4] * (n + 1)
        ps = []
        two_steps = []
        two_steps_set = set()

        # We can create 0 without choosing anything
        dp[0] = 1

        num = 1
        while num * num <= n:
            # Base Case: 1 step to reach this level
            dp[num * num] = 1
            ps.append(num * num)
            num += 1

        # This is for 2 steps
        for num1 in ps:
            for num2 in ps:
                if num1 + num2 <= n:
                    # Note: 3^2 + 4^2 = 5^2, so that's why we use min.
                    dp[num1 + num2] = min(dp[num1 + num2], 2)

                    if dp[num1 + num2] == 2 and num1 + num2 not in two_steps_set:
                        two_steps.append(num1 + num2)
                        two_steps_set.add(num1 + num2)

                else:
                    break

        # Sort the list
        two_steps.sort()

        # This is for 3 steps
        for p in ps:
            for t in two_steps:
                if p + t <= n:
                    dp[p + t] = min(dp[p + t], 3)

                else:
                    break

        return dp[n]
```

```python
solution = Solution()
```
