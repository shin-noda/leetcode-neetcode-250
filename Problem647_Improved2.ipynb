```python
class Solution:
    def countSubstrings(self, s):
        # Use Manacher's Algorithm

        # T: Transform string: e.g. "abc" -> "^#a#b#c#$"
        # ^ and $ are unique bounds so we don't have to check array limits
        T = '^#' + '#'.join(s) + '#$'
        n = len(T)

        # P: The Radius Array
        # We create an array P where P[i] stores the radius
        # (how far the palindrome extends from the center i)
        P = [0] * n

        # C: The Center of a palindrome
        C = 0

        # R: The Right boundary of the palindrome
        R = 0

        for i in range(1, n - 1):
            # i's mirror image on the left side of C
            # Equals: C - (i - C)
            i_mirror = 2 * C - i

            # If we are within the boundary R, copy the mirrored value
            if R > i:
                P[i] = min(R - i, P[i_mirror])

            # Attempt to expand centered at i
            while T[i - 1 - P[i]] == T[i + 1 + P[i]]:
                P[i] += 1

            # If the palindrome centered at i expands past R,
            # adjust the centre and right boundary to our new largest palindrome
            if i + P[i] > R:
                C = i
                R = i + P[i]

        # In the transformed string, the radius P[i] perfectly
        # equals the number of actual palindromes centered at i
        # (e.g., if radius is 3, it contributes exactly (3 + 1) // 2 actual palindromes)
        total_palindromes = sum((radius + 1) // 2 for radius in P)

        return total_palindromes
```
