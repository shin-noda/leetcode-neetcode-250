```python
class Solution:
    def countSubstrings(self, s):
        counts = 0
        n = len(s)
        dp = [[False] * n for _ in range(n)]

        # Base Case: length of 1
        for i in range(n):
            dp[i][i] = True
            counts += 1

        # Base Case: length of 2
        for i in range(n - 1):
            if s[i] == s[i + 1]:
                dp[i][i + 1] = True
                counts += 1

        for length in range(3, n + 1):
            for i in range(n + 1 - length):
                j = i + length - 1

                if s[i] == s[j] and dp[i + 1][j - 1]:
                    dp[i][j] = True
                    counts += 1

        return counts
```
