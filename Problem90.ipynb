```python
class Solution:
    def subsetsWithDup(self, nums):
        self.tmp = []
        self.answers = set()

        def backtrack(index):
            if index >= len(nums):
                self.answers.add(tuple(self.tmp))
                return

            for i in range(index, len(nums)):
                # If this number is the same as the previous one, we skip it
                if i != index and i >= 1 and nums[i] == nums[i - 1]:
                    continue

                # Choice 1: Choose
                self.tmp.append(nums[i])
                backtrack(i + 1)
                self.tmp.pop()

                # Choice 2: Not to choose
                backtrack(i + 1)


        # Sort the array
        nums.sort()

        backtrack(0)

        return [list(x) for x in self.answers]
```
