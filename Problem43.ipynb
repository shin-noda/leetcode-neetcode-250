```python
class Solution:
    def multiply(self, num1, num2):
        # Quick check
        if num1 == "0" or num2 == "0":
            return "0"

        # The maximum possible length of the product is len(num1) + len(num2)
        results = [0] * (len(num1) + len(num2))

        # Reverse loop to simulate right-to-left grade school multiplication
        for i in range(len(num1) - 1, -1, -1):
            for j in range(len(num2) - 1, -1, -1):
                # Multiply the single digits
                mul = (ord(num1[i]) - ord('0')) * (ord(num2[j]) - ord('0'))

                # Add to the current position
                p1, p2 = i + j, i + j + 1
                total = mul + results[p2]

                # Update the array with the carry and the remainder
                results[p2] = total % 10
                results[p1] += total // 10

        # Skip any leading zeros that weren't filled
        start = 0
        while start < len(results) and results[start] == 0:
            start += 1

        # Convert the integer array back to a string without using str(large_int)
        return "".join(map(str, results[start:]))
```

```python
solution = Solution()
```

```python
num1 = "2"
num2 = "3"

print(solution.multiply(num1, num2))
```

```python
num1 = "123"
num2 = "456"

print(solution.multiply(num1, num2))
```
