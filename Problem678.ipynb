```python
class Solution:
    def checkValidString(self, s):
        # Use the Range-Based Greedy Approach (Interval Tracking)
        # Minimum possible open '('
        left_min = 0

        # Maximum possible open '('
        left_max = 0

        for ch in s:
            if ch == '(':
                left_min += 1
                left_max += 1

            elif ch == ')':
                left_min -= 1
                left_max -= 1

            else:
                # '*' acts as ')'
                left_min -= 1

                # '*' acts as '('
                left_max += 1

            # Guardrail 1: Too many ')' encountered.
            if left_max < 0:
                return False

            # Guardrail 2: left_min can't be negative
            if left_min < 0:
                left_min = 0

        return left_min == 0
```

```python
solution = Solution()
```

```python
s = "()"

print(solution.checkValidString(s))
```

```python
s = "(*)"

print(solution.checkValidString(s))
```

```python
s = "(*))"

print(solution.checkValidString(s))
```

```python
s = ")("

print(solution.checkValidString(s))
```

```python
s = "())"

print(solution.checkValidString(s))
```
