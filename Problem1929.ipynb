```python
class Solution:
    def getConcatenation(self, nums):
        return nums * 2
```

```python
solution = Solution()
```

```python
nums = [1,2,1]

print(solution.getConcatenation(nums))
```

```python
nums = [1,3,2,1]

print(solution.getConcatenation(nums))
```
