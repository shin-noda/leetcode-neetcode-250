```python
class Solution:
    def stoneGame(self, piles):
        n = len(piles)

        dp = [[0] * n for _ in range(n)]

        for i in range(n):
            dp[i][i] = piles[i]

        for length in range(2, n + 1):
            for i in range(n - length + 1):
                j = i + length - 1

                # The maximum score difference (current player - other player)
                # that the current player can guarantee from piles i...j.
                # It's -dp, not +dp because that's your opponent's score.
                # Those -dp's are the opponent's future advantage.
                dp[i][j] = max(
                    piles[i] - dp[i + 1][j],
                    piles[j] - dp[i][j - 1]
                )

        return dp[0][n - 1] > 0
```

```python
solution = Solution()
```
