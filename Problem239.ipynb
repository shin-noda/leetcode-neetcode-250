```python
from collections import deque

class Solution:
    def maxSlidingWindow(self, nums, k):
        # Use Monotonic Stack (strictly descending order)
        # dq[0] is always the max of Sliding Window

        results = []

        # Store indices, not values
        dq = deque()

        for i, val in enumerate(nums):
            # Remove smaller values from the back (they'll never be the max)
            while dq and nums[dq[-1]] < val:
                dq.pop()

            dq.append(i)

            # Remove the front index if it's outside the window
            if dq[0] == i - k:
                dq.popleft()

            # Once the window is full size, the front is our max
            if i >= k - 1:
                results.append(nums[dq[0]])

        return results
```

```python
solution = Solution()
```

```python
nums = [1,3,-1,-3,5,3,6,7]
k = 3

print(solution.maxSlidingWindow(nums, k))
```
