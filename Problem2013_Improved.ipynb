```python
class DetectSquares:

    def __init__(self):
        # Create a graph which stores the number of points
        # e.g. self.adj[3][5] = 2 -> There are two (3, 5).
        self.adj = {}


    def add(self, point):
        x, y = point

        if x not in self.adj:
            self.adj[x] = {}

        if y not in self.adj[x]:
            self.adj[x][y] = 0

        self.adj[x][y] += 1


    def count(self, point):
        square_counts = 0
        x, y = point
        y_points = []
        point_set = set()

        if x in self.adj:
            y_points = list(self.adj[x])

        # We cannot form a square.
        if not y_points:
            return 0

        # Add vertical points
        for yp in y_points:
            point_set.add((x, yp))

        # At this point, if we find diagonal points,
        # we can form squares.
        while point_set:
            px, py = point_set.pop()

            if (x == px and y == py):
                if (px, py) in point_set:
                    point_set.remove((px, py))

                continue

            square_counts += self.getSquareCounts(x, y, px, py)

        return square_counts


    def getPointCounts(self, x, y):
        if x in self.adj and y in self.adj[x]:
            return self.adj[x][y]

        return 0


    def getSquareCounts(self, x, y, px, py):
        counts = p1x = p1y = p2x = p2y = p3x = p3y = p4x = p4y = ud = lr = 0
        p_counts = self.getPointCounts(px, py)

        # It's guaranteed that one of two is 0
        edge = max(abs(x - px), abs(y - py))

        if x == px:
            # Left and Right points
            p1x, p1y = x - edge, y
            p2x, p2y = x + edge, y

            if y > py:
                ud = -1

            else:
                ud = 1

            # We check if there are squares at either:
            # - bottom-left and bottom-right or
            # - upper-left and upper-right
            p3x, p3y = x - edge, y + edge * ud
            p4x, p4y = x + edge, y + edge * ud

        # This is for y == py
        else:
            # Upper and Bottom points
            p1x, p1y = x, y + edge
            p2x, p2y = x, y - edge

            if x > px:
                lr = -1

            else:
                lr = 1

            # We check if there are squares at either:
            # - upper-left and bottom-left.
            # - upper-right and bottom-right
            p3x, p3y = x + edge * lr, y + edge
            p4x, p4y = x + edge * lr, y - edge

        p1_counts = self.getPointCounts(p1x, p1y)
        p2_counts = self.getPointCounts(p2x, p2y)
        p3_counts = self.getPointCounts(p3x, p3y)
        p4_counts = self.getPointCounts(p4x, p4y)

        square_counts_1 = p_counts * p1_counts * p3_counts
        square_counts_2 = p_counts * p2_counts * p4_counts

        counts += square_counts_1 + square_counts_2

        return counts
```
