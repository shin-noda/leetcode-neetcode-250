```python
class DetectSquares:

    def __init__(self):
        # Create a graph which stores the number of points
        # e.g. self.adj[3][5] = 2 -> There are two (3, 5).
        self.adj = {}


    def add(self, point):
        x, y = point
        if x not in self.adj:
            self.adj[x] = {}

        if y not in self.adj[x]:
            self.adj[x][y] = 0

        self.adj[x][y] += 1


    def count(self, point):
        square_counts = 0
        x, y = point
        y_points = []
        point_set = set()

        if x in self.adj:
            y_points = list(self.adj[x])

        # We cannot form a square.
        if not y_points:
            return 0

        # Add vertical points
        for yp in y_points:
            point_set.add((x, yp))

        # At this point, if we find diagonal points,
        # we can form squares.
        while point_set:
            px, py = point_set.pop()

            if y == py:
                continue

            square_counts += self.getSquareCounts(x, y, px, py)

        return square_counts


    def getPointCounts(self, x, y):
        if x in self.adj and y in self.adj[x]:
            return self.adj[x][y]

        return 0


    def getSquareCounts(self, x, y, p0x, p0y):
        p1x = p1y = p2x = p2y = p3x = p3y = p4x = p4y = 0

        # It's guaranteed that one of two is 0
        edge = abs(y - p0y)

        # Left and Right points
        p1x, p1y = x - edge, y
        p2x, p2y = x + edge, y

        # We check if there are squares at either:
        # - bottom-left and bottom-right or
        # - upper-left and upper-right
        p3x, p3y = x - edge, p0y
        p4x, p4y = x + edge, p0y

        p0_counts = self.getPointCounts(p0x, p0y)
        p1_counts = self.getPointCounts(p1x, p1y)
        p2_counts = self.getPointCounts(p2x, p2y)
        p3_counts = self.getPointCounts(p3x, p3y)
        p4_counts = self.getPointCounts(p4x, p4y)

        return (p0_counts * p1_counts * p3_counts) + (p0_counts * p2_counts * p4_counts)
```
