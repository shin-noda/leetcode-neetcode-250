```python
class DetectSquares:

    def __init__(self):
        # Create a graph which stores the number of points
        # e.g. self.adj[3][5] = 2 -> There are two (3, 5).
        self.adj = {}


    def getPointCounts(self, x, y):
        if x in self.adj and y in self.adj[x]:
            return self.adj[x][y]

        return 0


    def add(self, point):
        x, y = point
        if x not in self.adj:
            self.adj[x] = {}

        if y not in self.adj[x]:
            self.adj[x][y] = 0

        self.adj[x][y] += 1


    def count(self, point):
        square_counts = 0
        x, y = point

        # Guardrail
        if x not in self.adj:
            return 0

        # At this point, if we find diagonal points,
        # we can form squares.
        for py in self.adj[x]:
            if y == py:
                continue

            edge = abs(y - py)

            square_counts += (
                (
                    self.getPointCounts(x, py) *
                    self.getPointCounts(x - edge, y) *
                    self.getPointCounts(x - edge, py)
                ) +
                (
                    self.getPointCounts(x, py) *
                    self.getPointCounts(x + edge, y) *
                    self.getPointCounts(x + edge, py)
                )
            )

        return square_counts
```
