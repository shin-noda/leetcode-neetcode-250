```python
from collections import deque

class Solution:
    def search(self, nums, target):
        # Brute force approach:
        # Find the max value and then, find the "cliff" so that we can find
        # how much the array was rotated.
        # Then, store the original array and perform Binary Search.
        # It's O(N).

        # For the follow-up question, the answer is yes.
        # Because since there are duplicates, there are cases like
        # [2, 2, 2, ..., 2 (mid), 3, 10, 0, 1, ... 2] or
        # [2, 3, 6, 10,0, ..., 2(mid), 2, 2, .. ,2].
        # We cannot just tell where to search next because there are "flat" spaces
        # by only relying on left, mid, and right pointers.

        # Implement the brute force solution: O(N)

        # Check if there is only one element in the array
        if len(nums) == 1:
            return nums[0] == target

        rotation = 0
        for i in range(1, len(nums)):
            # We found the "cliff"
            if nums[i - 1] > nums[i]:
                rotation = i

        dq = deque(nums)
        for _ in range(rotation):
            dq.append(dq.popleft())

        left = 0
        right = len(dq) - 1

        while left <= right:
            mid = (left + right) // 2

            if dq[mid] == target:
                return True

            elif dq[mid] < target:
                left = mid + 1

            else:
                right = mid - 1

        return False
```

```python
solution = Solution()
```

```python
nums = [2,5,6,0,0,1,2]
target = 0

print(solution.search(nums, target))
```

```python
nums = [2,5,6,0,0,1,2]
target = 3

print(solution.search(nums, target))
```
