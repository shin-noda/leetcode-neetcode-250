```python
class Solution:
    def isAlienSorted(self, words, order):
        # O(1) space since it's always exactly 26 keys
        alien_dict = {char: idx for idx, char in enumerate(order)}

        # Compare adjacent words
        for i in range(len(words) - 1):
            w1, w2 = words[i], words[i + 1]

            min_len = min(len(w1), len(w2))

            # It catches the edge case where "apple" comes before "app"
            if len(w1) > len(w2) and w1[:min_len] == w2[:min_len]:
                return False

            # Check character by character up to the shorter word's length
            for j in range(min_len):
                if w1[j] != w2[j]:
                    if alien_dict[w1[j]] > alien_dict[w2[j]]:
                        return False

                    break

        return True
```
