```python
class Solution:
    def findInMountainArray(self, target, mountainArr):
        def binarySearch(left, right):
            while left <= right:
                mid = (left + right) // 2
                mid_val = mountainArr.get(mid)

                if mid_val == target:
                    return mid

                elif mid_val < target:
                    left = mid + 1

                else:
                    right = mid - 1

            return -1


        def binarySearchReversed(left, right):
            while left <= right:
                mid = (left + right) // 2
                mid_val = mountainArr.get(mid)

                if mid_val == target:
                    return mid

                elif mid_val > target:
                    left = mid + 1

                else:
                    right = mid - 1

            return -1


        # Note: the peak exists between 0 < i < len(arr) - 1
        # In other words, arr like
        # [5, 4, ..., 1] or [1, 2, ..., 5]
        # are invalid because the peak exists at the beginning or the end

        # We find the peak first

        n = mountainArr.length()

        peak = 0
        left = 0
        right = n - 1

        # Worst Case: O(logN): log2(10^4) = 14, so 14 * 3 = 42 get calls
        while left <= right:
            mid = (left + right) // 2

            # We have a limited number of .get() calls, so we use variables
            # to reduce the number of calls.
            mid_val = mountainArr.get(mid)

            mid_minus = mid_val
            if mid - 1 >= 0:
                mid_minus = mountainArr.get(mid - 1)


            mid_plus = mid_val
            if mid + 1 <= n - 1:
                mid_plus = mountainArr.get(mid + 1)

            # We check the slope left and right next to the mid
            if mid_val > mid_minus and mid_val > mid_plus:
                peak = mid
                if mid_val == target:
                    return mid

                break

            elif mid_val < mid_plus:
                # We go to the right side because the right side is uphill
                left = mid + 1

            else:
                # We go to the left side because the left side is uphill
                right = mid - 1

        # Next, we search the left to the peak
        left = 0
        right = peak - 1
        result = binarySearch(left, right)

        if result != -1:
            return result

        # We check the right to the peak
        left = peak + 1
        right = n - 1

        return binarySearchReversed(left, right)
```
