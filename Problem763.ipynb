```python
class Solution:
    def partitionLabels(self, s):
        # Idea:
        # Marge overlapping intervals

        ch_map = {}
        for i in range(len(s)):
            c = s[i]

            if c not in ch_map:
                ch_map[c] = []

            ch_map[c].append(i)

        partitions = []
        keys = list(ch_map.keys())
        curr_s, curr_e = ch_map[keys[0]][0], ch_map[keys[0]][-1]

        for i in range(1, len(keys)):
            key = keys[i]
            start, end = ch_map[key][0], ch_map[key][-1]

            # It's a new interval
            if start > curr_e:
                partitions.append(curr_e - curr_s + 1)

                curr_s = start
                curr_e = end

            else:
                curr_s = min(curr_s, start)
                curr_e = max(curr_e, end)

        # Add the last interval
        partitions.append(curr_e - curr_s + 1)

        return partitions
```

```python
solution = Solution()
```

```python
s = "ababcbacadefegdehijhklij"

print(solution.partitionLabels(s))
```
