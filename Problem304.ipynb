```python
class NumMatrix:

    def __init__(self, matrix):
        # Idea: we calculate prefix sum

        rows, cols = len(matrix), len(matrix[0])

        # Initialize prefix sum matrix with padding
        # We add a fake border fo zeros, so
        # the top row and the loeft column are always 0
        # -> matrix[0][0] -> Prefix[1][1]
        # If we don't pad, we have to check out of bounds errors.
        self.prefix = [[0] * (cols + 1) for _ in range(rows + 1)]

        for r in range(rows):
            for c in range(cols):
                # Each prefix cell stores a top-left accumulated sum.
                # Current prefix = current val + top + left - top_left_diagonal
                self.prefix[r + 1][c + 1] = (
                    matrix[r][c] +
                    self.prefix[r][c + 1] +
                    self.prefix[r + 1][c] -
                    self.prefix[r][c]
                )


    def sumRegion(self, row1, col1, row2, col2):
        # Use the inclusion-excusion principle
        return (
            self.prefix[row2 + 1][col2 + 1] -
            self.prefix[row1][col2 + 1] -
            self.prefix[row2 + 1][col1] +
            self.prefix[row1][col1]
        )
```
