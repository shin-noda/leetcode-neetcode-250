```python
class Solution:
    def subarraySum(self, nums, k):
        # We use:
        # current sum - k = target previous prefix

        count = 0
        current_sum = 0

        # Map to store prefix sum frequencies
        # This is our safety net for when a subarray starting right from
        # the very beginning (index 0) adds up to k
        prefix_counts = {0: 1}

        for num in nums:
            current_sum += num

            # If (current_sum - k) was seen before, it means a
            # subarray exists that sums to k
            diff = current_sum - k
            if diff in prefix_counts:
                count += prefix_counts[diff]

            # Record this prefix sum in the map
            prefix_counts[current_sum] = prefix_counts.get(current_sum, 0) + 1

        return count
```

```python
solution = Solution()
```

```python
nums = [1,1,1]
k = 2

print(solution.subarraySum(nums, k))
```
