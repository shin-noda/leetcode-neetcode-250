```python
class Solution:
    def majorityElement(self, nums):
        # Use Boyer-Moore Voting Algorithm

        # Since we're looking for numbers which apprear more than n/3 times,
        # there are at most 2 candidates.
        # Suppose there were 3 different numbers that each appeared more than n/3 times.
        # The would mean:
        # First number appears > n/3
        # Second number appears > n/3
        # Third number appears > n /3
        # Add them together:
        # more than n/3 + more than n/3 + more than n/3 = more than n
        # Thus, this is impossible.
        # Therefore, there can only be:
        # 0 answers, 1 answer, or 2 answers.

        if not nums:
            return []

        # 1st Pass: Find the two most sturdy candidates
        # We only keep track of two candidates.
        candidate1, count1 = None, 0
        candidate2, count2 = None, 0

        for num in nums:
            if candidate1 == num:
                count1 += 1

            elif candidate2 == num:
                count2 += 1

            elif count1 == 0:
                candidate1, count1 = num, 1

            elif count2 == 0:
                candidate2, count2 = num, 1

            else:
                # The three-way cancel
                # Current num, candidate1, and candiate2 are all different
                count1 -= 1
                count2 -= 1

        # 2nd Pass: Verify the candidates
        # The algorithm only guarantees these are the best candidates
        # not that they actually meet the > n/3 threshhold
        results = []

        # Note: the loop is constant because it's O(2).
        # This this loop's time complexity is:
        # .count -> O(N), O(2 * N) -> O(N)
        for cand in [candidate1, candidate2]:
            if cand is not None and nums.count(cand) > len(nums) // 3:
                results.append(cand)

        return results
```

```python
solution = Solution()
```

```python
nums = [3,2,3]

print(solution.majorityElement(nums))
```
