```python
class Solution:
    def isMatch(self, s, p):
        # Idea:
        # Maybe I cannot solve this problem Greedly :(

        # Maybe I need to process 's' backwards???

        # Anoter idea:
        # 2D dp[i][j] where dp is len(s) * len(p) matrix
        # If dp[-1][-1] is true, it's true.


        # Helper function
        def useMemo():
            i, j = -1, -1

            while memo:
                tmp_i, tmp_j = memo.pop()

                if (tmp_i, tmp_j) not in used:
                    i = tmp_i
                    j = tmp_j

                    # Add it to the used list
                    used.add((i, j))
                    break

            return i, j


        # Helper function 2
        def createMemo(i, j, ch):
            j += 2

            if ch == '.':
                for index in range(i, len(s)):
                    memo.add((index, j))

                # We check if we can find the matching pattern
                if canExhaustPattern(j):
                    memo.add((len(s), len(p)))

            else:
                # Take a snap shot if it's valid.
                while i < len(s) and ch == s[i]:
                    # We don't store invalid indices
                    # We store (i, j) = (n, m) because that means it's a valid pattern
                    if i == len(s) and j == len(p) or i < len(s) and j < len(p):
                        memo.add((i, j))

                    i += 1

            return i, j


        # Helper function 3
        def canExhaustPattern(j):
            # This is a function for dealing with the case that
            # the index i can reach at the end of 's' but
            # the index j hasn't reanched the end of 'p'

            # The end of p has to be "*" because if
            # it's an English letter or '.', there is no character
            # left in s to match with.
            if p[-1] != '*':
                return False

            # We proceed 'j' by 2
            # The rest of 'j' has to be a combination of
            # 'One character' + '*'.
            # This is the only combination that can
            # match with zero characters.
            while j < len(p):
                if j + 1 >= len(p) or p[j + 1] != '*':
                    return False

                j += 2

            # Check if 'j' could reach the end of p
            return j >= len(p)


        # Quick Check
        # ".*" means any expression
        if p == ".*":
            return True

        n = len(s)
        m = len(p)

        # Index of s
        i = 0

        # Index of p
        j = 0

        # Replay function.
        memo = set()

        # We store replayed (i, j) to avoid infinite loop
        used = set()

        while i < n or j < m:
            # 's' or 'p' reached at the end but not both at the same time
            if i >= n or j >= m:
                if i >= n and canExhaustPattern(j):
                    return True

                if not memo:
                    return False

                else:
                    i, j = useMemo()

                    if i == -1 and j == -1:
                        return False

            ch = p[j]

            # Check if the next character is Kleene Star or not
            if j + 1 < m and p[j + 1] == '*':
                # We store the valid combinations
                i, j = createMemo(i, j, ch)

                if (n, m) in memo:
                    return True

            # It's not a star.
            else:
                # Check if ch matches with s[i]
                # If ch is the wildcard, we can proceed
                if ch != '.' and ch != s[i]:
                    i, j = useMemo()

                    if i == -1 and j == -1:
                        return False

                else:
                    i += 1
                    j += 1

        return True
```
