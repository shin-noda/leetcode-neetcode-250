```python
class MyHashSet:

    def __init__(self):
        self.my_list = []


    def add(self, key):
        if key not in self.my_list:
            self.my_list.append(key)


    def remove(self, key):
        if key in self.my_list:
            for i in range(len(self.my_list)):
                item = self.my_list[i]

                if item == key:
                    self.my_list.pop(i)
                    break


    def contains(self, key):
        if key in self.my_list:
            return True

        return False
```
