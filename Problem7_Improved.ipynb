```python
class Solution:
    def reverse(self, x):
        # Define 32-bit signed integer boundary
        MIN_INT, MAX_INT = -2**31, 2**31 - 1

        result = 0
        while x != 0:
            if x < 0:
                digit = x % -10

                # Python's // rounds down
                x = int(x / 10)

            else:
                digit = x % 10
                x //= 10

            # Check overflow (Positive side)
            if result > MAX_INT // 10 or (result == MAX_INT // 10 and digit > 7):
                return 0

            # Check underflow (Negative side)
            if result < int(MIN_INT / 10) or (result == int(MIN_INT / 10) and digit < -8):
                return 0

            result = (result * 10) + digit

        return result
```

```python
solution = Solution()
```

```python
x = 123

print(solution.reverse(x))
```
