```python
class Solution:
    def splitArray(self, nums, k):
        # Check if k is 1
        if k == 1:
            return sum(nums)

        # Check if k has the same value as the length of nums
        if k == len(nums):
            return max(nums)

        # The range of min_sum is: sum(nums) // k <= min_sum <= sum(nums)
        left = sum(nums) // k
        right = sum(nums)
        min_sum = right

        while left <= right:
            # This is the threshhold, so we put numbers
            # until it reaches mid when groups is less than k
            mid = (left + right) // 2

            groups = 1
            zeros = 0
            curr = 0
            max_val = 0

            for num in nums:
                if num == 0:
                    zeros += 1

                if groups < k:
                    if curr + num <= mid:
                        curr += num
                        max_val = max(max_val, curr)

                    # The current group is gonna overshoot
                    else:
                        # Reset the curr
                        curr = num
                        max_val = max(max_val, curr)

                        groups += 1

                # We simply add the value: we cannot add another group
                else:
                    curr += num
                    max_val = max(max_val, curr)

            # groups is at most k
            # Note: we need to account for zeros
            if groups == k or (groups < k and groups + zeros >= k):
                # The potential value
                min_sum = min(min_sum, max_val)

                if max_val <= mid:
                    # Maybe we can put more values in each group, so
                    # shrink the value of mid
                    right = mid - 1

                else:
                    # Mid was too small, so we need to balance more
                    left = mid + 1

            # Number of groups is too small
            else:
                # Try to find a better value
                right = mid - 1

        return min_sum
```
