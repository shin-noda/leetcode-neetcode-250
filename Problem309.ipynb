```python
class Solution:
    def maxProfit(self, prices):
        # Initialize our states for the start
        # Float("-inf") ensures we must actually buy a stock to enter this state
        buying = float("-inf")
        selling = 0
        cooldown = 0

        for price in prices:
            # We must use temporary variables because the states update simultaneously
            next_buying = max(buying, cooldown - price)
            next_selling = buying + price
            next_cooldown = max(cooldown, selling)

            # Update states for the next day
            buying = next_buying
            selling = next_selling
            cooldown = next_cooldown

        # At the end, maximum profit will never be in the 'buying' state
        # It's either we just sold, or we are chilling in cooldown
        return max(selling, cooldown)
```
