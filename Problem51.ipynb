```python
class Solution:
    def solveNQueens(self, n):
        # Note: Vertical Lines, Left Diagonal Lines and Right Diagonal Lines
        vs = set()
        lds = set()
        rds = set()
        queens = set()
        results = []
        board = [["."] * n for _ in range(n)]


        def backtrack(r):
            # Base Case
            if r >= n:
                compressed_board = []
                for row in board:
                    compressed_board.append("".join(row))

                results.append(list(compressed_board))

                return

            for c in range(n):
                # Pruning
                if (r, c) in vs or (r, c) in lds or (r, c) in rds:
                    continue

                # Choose
                board[r][c] = 'Q'

                for i in range(r + 1, n):
                    vs.add((i, c))

                i = r + 1
                j = c - 1
                while i <= n - 1 and 0 <= j:
                    lds.add((i, j))

                    i += 1
                    j -= 1

                i = r + 1
                j = c + 1
                while i <= n - 1 and j <= n - 1:
                    rds.add((i, j))

                    i += 1
                    j += 1

                # Explore
                backtrack(r + 1)

                # Unchoose
                board[r][c] = '.'

                for i in range(r + 1, n):
                    vs.remove((i, c))

                i = r + 1
                j = c - 1
                while i <= n - 1 and 0 <= j:
                    lds.remove((i, j))

                    i += 1
                    j -= 1

                i = r + 1
                j = c + 1
                while i <= n - 1 and j <= n - 1:
                    rds.remove((i, j))

                    i += 1
                    j += 1

        backtrack(0)

        return results
```
