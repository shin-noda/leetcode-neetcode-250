```python
class Solution:
    def characterReplacement(self, s, k):
        curr_streak = 0
        max_streak = 1

        # We always flip the leftmost char to match the rightmost char.
        flips = 0

        left = 0
        prev = s[0]
        for right in range(1, len(s)):
            if s[right] != prev:
                flips += 1

                # When number of flips is greater than k, we
                # instantly shrink the window size to zero and expand
                # to the left from the current right position
                if flips > k:
                    left = right
                    flips = 0

                    # We change the char which will be compared
                    prev = s[right]

            # We expand the left window
            while left > 0:
                if s[left - 1] != s[right]:
                    # We cannot expand the leftside anymore
                    if flips + 1 > k:
                        break

                    else:
                        flips += 1

                left -= 1

            curr_streak = right - left + 1
            max_streak = max(max_streak, curr_streak)

        return max_streak
```

```python
solution = Solution()
```
