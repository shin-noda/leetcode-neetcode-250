```python
class Solution:
    def findTargetSumWays(self, nums, target):
        # Idea:
        # This problem is a variation of Last Stone Weight II.
        # This is a 0/1 Knapsack problem.
        # Think about putting all items in two knapsacks.
        # Knapsack A for positive and Knapsack B for negative numbers.

        # Implement the subset-sum solution.

        # We split nums into two groups.
        # + group (P) -> P: sum of numbers with +
        # - group (N) -> N: sum of numbers with -
        #
        # Then the equation becomes:
        # P - N = target
        # Also, all numbers are used:
        # P + N = total_sum
        #
        # We combine the two equations
        # P - N = target
        # P + N = total_sum
        # Add them:
        # 2P = target + total_sum
        # Thus,
        # P = (target + total_sum) / 2
        #
        # Now, instead of assigning + and =, we ask:
        # How many subsets of nums have sum = P?
        # -> Subset Sum Counting

        total_sum = sum(nums)

        # Mathematical impossibility checks
        # For the part: if target is bigger than total possible sum, it's impossible.
        # For the second part: if (target + total_sum) / 2 is not an integer, it's impossible.
        if abs(target) > total_sum or (target + total_sum) % 2 != 0:
            return 0

        # Our new subset target sum
        # This subset_target is P
        subset_target = (target + total_sum) // 2

        # dp[i] stores the number of ways to reach sum i
        dp = [0] * (subset_target + 1)

        # Base case: 1 way to make a sum of 0 (the empty set)
        dp[0] = 1

        for num in nums:
            # Traverse backwards to use the value from the previous row
            # and prevent using the same element multiple times
            # We always use values from the previous state, not the updated one.
            for j in range(subset_target, num - 1, -1):
                # We want number of ways to make sum j.
                dp[j] += dp[j - num]

        return dp[subset_target]
```

```python
solution = Solution()
```

```python
nums = [1,1,1,1,1]
target = 3

print(solution.findTargetSumWays(nums, target))
```

```python
nums = [1]
target = 1

print(solution.findTargetSumWays(nums, target))
```
