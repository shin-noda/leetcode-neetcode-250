```python
class Solution:
    def maxProduct(self, nums):
        # Idea:
        # When there is a 0, it resets the product.
        # When there is a negative number, it flips.
        # When there is a second negative nuber, it flips again and behaves like
        # a positive number

        # e.g.
        # [1,2,3,-4,5,6,7,8,9,0,11,12,13,14,-15,16,17,18]
        # -> the max subrray is 11 * 12 * 13 * 14

        n = len(nums)

        curr = nums[0]
        max_val = max(nums)

        # pos and neg keep track of the value when the sign of curr is fliped
        pos = 0
        neg = 0

        for i in range(1, n):
            if curr == 0:
                curr = nums[i]

            else:
                curr_pos = pos
                curr_neg = neg

                pos = max(nums[i], curr * nums[i], curr_pos * nums[i], curr_neg * nums[i])
                neg = min(nums[i], curr * nums[i], curr_neg * nums[i], curr_pos * nums[i])

                curr *= nums[i]

            max_val = max(max_val, curr, pos)

        return max_val
```

```python
solution = Solution()
```
