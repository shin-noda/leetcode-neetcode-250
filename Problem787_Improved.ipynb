```python
class Solution:
    def findCheapestPrice(self, n, flights, src, dst, k):
        # Bellman-Ford / Level-by-level BFS approach
        costs = [float('inf')] * n
        costs[src] = 0

        # We can take at most k + 1 flights (edges)
        for _ in range(k + 1):
            # Clone current state to avoid cascading updates.
            temp_costs = list(costs)

            for u, v, price in flights:
                if costs[u] == float('inf'):
                    continue

                if costs[u] + price < temp_costs[v]:
                    temp_costs[v] = costs[u] + price

            costs = temp_costs

        if costs[dst] != float('inf'):
            return costs[dst]

        return -1
```
