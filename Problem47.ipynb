```python
class Solution:
    def permuteUnique(self, nums):
        self.results = set()

        def backtrack(i, j):
            # Base case
            if i >= len(nums) - 1:
                return

            # Calculate new i and j.
            # We swap i-th and j-th elements later, so we need to preserve current i and j.
            new_i = i
            new_j = j + 1
            if new_j >= len(nums):
                new_i += 1
                new_j = new_i + 1

            # Choice 1: Swap (we swap with previous i and j to backtrack later)
            nums[i], nums[j] = nums[j], nums[i]

            self.results.add(tuple(nums))

            # We skip duplicates
            if nums[i] != nums[j]:
                backtrack(new_i, new_j)

            # Backtrack: swap back again
            nums[i], nums[j] = nums[j], nums[i]

            # Choice 2: Not to swap
            backtrack(new_i, new_j)

        self.results.add(tuple(nums))
        backtrack(0, 0)

        return [list(x) for x in self.results]
```
