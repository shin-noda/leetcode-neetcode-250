```python
class Solution:
    def subsetXORSum(self, nums):
        # Idea:
        # We compute the bitwise OR of the entire array in a single pass, and
        # then shift it left by n - 1 to multiply by 2^(n - 1)

        bits = 0
        for x in nums:
            bits |= x

        return bits << (len(nums) - 1)
```
