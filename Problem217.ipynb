```python
class Solution:
    def containsDuplicate(self, nums):
        return len(nums) != len(set(nums))
```

```python
solution = Solution()
```

```python
nums = [1,2,3,1]

print(solution.containsDuplicate(nums))
```

```python
nums = [1,2,3,4]

print(solution.containsDuplicate(nums))
```

```python
nums = [1,1,1,3,3,4,3,2,4,2]

print(solution.containsDuplicate(nums))
```

```python
nums = [1]

print(solution.containsDuplicate(nums))
```
