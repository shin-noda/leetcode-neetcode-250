```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
class Solution:
    def diameterOfBinaryTree(self, root):
        self.diameter = 0

        def dfs(node):
            if not node:
                return 0

            # Ask the children for their depth
            left = dfs(node.left)
            right = dfs(node.right)

            # Update the global diameter if the path though this node is the longest seen so far
            self.diameter = max(self.diameter, left + right)

            # Return the depth of the current subtree back up to the parent
            return 1 + max(left, right)

        # start the search
        dfs(root)

        return self.diameter
```
