```python
import heapq

class Solution:
    def minInterval(self, intervals, queries):
        # Sort intervals by their start point
        intervals.sort(key=lambda x: x[0])

        # Pair queries with their original incides and sort them
        sorted_queries = sorted((q, idx) for idx, q in enumerate(queries))

        # Result array to map answers to map answers back to original query incides
        results = [-1] * len(queries)

        # Min-heap stores: (size_of_interval, right_boundary)
        min_heap = []

        # Pointer for intervals
        i = 0

        # Process each query in increasing order
        for q, original_idx in sorted_queries:
            # Add all intervals that start before or at the current query
            while i < len(intervals) and intervals[i][0] <= q:
                left, right = intervals[i]
                size = right - left + 1
                heapq.heappush(min_heap, (size, right))
                i += 1

            # Remove intervals from the top of the heap that end before the query
            while min_heap and min_heap[0][1] < q:
                heapq.heappop(min_heap)

            # The top of the heap is now our smallest valid interval
            if min_heap:
                results[original_idx] = min_heap[0][0]

        return results
```

```python
solution = Solution()
```

```python
intervals = [[1,4],[2,4],[3,6],[4,4]]
queries = [2,3,4,5]

print(solution.minInterval(intervals, queries))
```
