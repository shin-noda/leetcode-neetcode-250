```python
import heapq

class Twitter:
    def __init__(self):
        self.time = 0

        # userId -> list of (neg_time, tweetId)
        self.tweets = {}

        # userId -> set of followeeIds
        self.following = {}


    def postTweet(self, userId, tweetId):
        if userId not in self.tweets:
            self.tweets[userId] = []

        if userId not in self.following:
            self.following[userId] = set()

        # Decrement time so larger timestamps become smaller numbers
        self.time -= 1
        self.tweets[userId].append((self.time, tweetId))


    def getNewsFeed(self, userId):
        if userId not in self.following:
            return []

        # Combine followed users and the user themselves
        all_users = self.following[userId] | {userId}
        heap = []

        # Push the most recent tweet from each user's timeline into the heap
        # Format: (neg_time, tweetId, followeeId, next_index_to_read)
        for f_id in all_users:
            if f_id in self.tweets and self.tweets[f_id]:
                # Start at the tail (newest)
                idx = len(self.tweets[f_id]) - 1
                neg_time, t_id = self.tweets[f_id][idx]
                heap.append((neg_time, t_id, f_id, idx - 1))

        # O(K) heapify where K is the number of users being tracked
        heapq.heapify(heap)

        results = []
        # Pull up to 10 of the absolute newest tweets across all streams
        while heap and len(results) < 10:
            net_time, t_id, f_id, next_idx = heapq.heappop(heap)
            results.append(t_id)

            # If that timeline still has older tweets, push the next one down the line
            if next_idx >= 0:
                nxt_time, nxt_id = self.tweets[f_id][next_idx]
                heapq.heappush(heap, (nxt_time, nxt_id, f_id, next_idx - 1))

        return results


    def follow(self, followerId, followeeId):
        if followerId == followeeId:
            return

        if followerId not in self.following:
            self.following[followerId] = set()

        if followeeId not in self.following:
            self.following[followeeId] = set()

        self.following[followerId].add(followeeId)


    def unfollow(self, followerId, followeeId):
        if followerId in self.following and followeeId in self.following[followerId]:
            self.following[followerId].remove(followeeId)
```
