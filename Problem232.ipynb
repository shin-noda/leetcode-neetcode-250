```python
class MyQueue:

    def __init__(self):
        # Those lists only behave like stacks
        self.stack = []
        self.tmp = []

    def push(self, x):
        # First, put everything into tmp
        n = len(self.stack)
        for _ in range(n):
            self.tmp.append(self.stack.pop())

        # Add a new value to tmp
        self.tmp.append(x)

        # Now, put everything into stack
        n = len(self.tmp)
        for _ in range(n):
            self.stack.append(self.tmp.pop())


    def pop(self):
        # O(1) operation
        return self.stack.pop()


    def peek(self):
        # O(1) operation
        return self.stack[-1]


    def empty(self):
        return len(self.stack) == 0
```
