```python
class Solution:
    def sortColors(self, nums):
        """
        Do not return anything, modify nums in-place instead.
        """

        # Use the Dutch National Flag algorithm.
        # The Core Logic: Three Pointers
        # Use three pointers to partition the array into four zones:
        # 1. Low: Everything before low is a 0
        # 2. Mid: Everything between low and mid is a 1
        # 3. Hight: Everything after high is a 2
        # 4. Unprocessed: The space between mid and high

        low = 0
        mid = 0
        high = len(nums) - 1

        while mid <= high:
            if nums[mid] == 0:
                # Swap mid with low to move the 0 to the front
                nums[low], nums[mid] = nums[mid], nums[low]
                low += 1
                mid += 1

            elif nums[mid] == 1:
                # It's in the right place (the middle zone), just move on
                mid += 1

            # nums[mid] == 2
            else:
                # We do not increment mid here.
                # We don't know what we just swapped from the end,
                # so we need to evaluate the new nums[mid] in the next iteration

                # Swap mid with high to move the 2 to the end
                nums[high], nums[mid] = nums[mid], nums[high]
                high -= 1
```

```python
solution = Solution()
```

```python
nums = [2,0,2,1,1,0]

print(solution.sortColors(nums))
```
