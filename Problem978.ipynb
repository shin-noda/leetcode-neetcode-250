```python
class Solution:
    def maxTurbulenceSize(self, arr):
        # Quick check
        if not arr:
            return 0

        # Up: +1, Down: -1, Neutral: 0
        ud = 0

        n = len(arr)
        max_turb = 1
        streak = 1

        for i in range(1, n):
            is_up = arr[i] > arr[i - 1]
            is_down = arr[i] < arr[i - 1]

            # Continue the current streak
            if (ud == 1 and is_down) or (ud == -1 and is_up):
                streak += 1
                ud *= -1

            # Start a new streak with moving upwards first
            elif is_up:
                streak = 2
                ud = 1

            # Start a new streak with moving downwardss first
            elif is_down:
                streak = 2
                ud = -1

            # No turbulance
            else:
                streak = 1
                ud = 0

            max_turb = max(max_turb, streak)

        return max_turb
```

```python
solution = Solution()
```
