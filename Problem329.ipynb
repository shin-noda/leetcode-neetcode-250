```python
class Solution:
    def longestIncreasingPath(self, matrix):
        rows, cols = len(matrix), len(matrix[0])
        dp = [[-1] * cols for _ in range(rows)]
        self.max_val = 0


        def dfs(r, c, prev, streak):
            # Base Case
            if (
                not (0 <= r < rows) or
                not (0 <= c < cols) or
                prev >= matrix[r][c] or
                streak < dp[r][c]
            ):
                return

            # Update the dp table
            streak += 1
            dp[r][c] = streak
            self.max_val = max(self.max_val, streak)

            # Up, Down, Left, and Right
            dfs(r - 1, c, matrix[r][c], streak)
            dfs(r + 1, c, matrix[r][c], streak)
            dfs(r, c - 1, matrix[r][c], streak)
            dfs(r, c + 1, matrix[r][c], streak)


        for r in range(rows):
            for c in range(cols):
                dfs(r, c, -1, 0)

        return self.max_val
```

```python
solution = Solution()
```

```python
matrix = [[9,9,4],[6,6,8],[2,1,1]]

print(solution.longestIncreasingPath(matrix))
```
