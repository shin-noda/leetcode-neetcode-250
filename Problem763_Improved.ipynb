```python
class Solution:
    def partitionLabels(self, s):
        # Record the last occurrence of each character
        last_occurrence = {ch: i for i, ch in enumerate(s)}

        partitions = []
        start = 0
        end = 0

        # Walk through the string and greedily find partition cuts
        for i, ch in enumerate(s):
            # Extend our current partition's end to include the last occurrence of 'ch'
            end = max(end, last_occurrence[ch])

            # If we've reached the end of the current partition
            if i == end:
                partitions.append(end - start + 1)

                # Next partition starts at the next index
                start = i + 1

        return partitions
```

```python
solution = Solution()
```

```python
s = "ababcbacadefegdehijhklij"

print(solution.partitionLabels(s))
```
