```python
import heapq

class Solution:
    def kClosest(self, points, k):
        # Check if elements of points are less than or equal to k.
        if len(points) <= k:
            return points

        max_heap = []
        for i in range(k):
            d = pow(pow(points[i][0], 2) + pow(points[i][1], 2), 0.5)
            max_heap.append((-d, points[i][0], points[i][1]))

        heapq.heapify(max_heap)

        for i in range(k, len(points)):
            d = pow(pow(points[i][0], 2) + pow(points[i][1], 2), 0.5)

            if d < -max_heap[0][0]:
                heapq.heappushpop(max_heap, (-d, points[i][0], points[i][1]))

        results = []
        for i in range(len(max_heap)):
            results.append([max_heap[i][1], max_heap[i][2]])

        return results
```
