```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
class Solution:
    def insertIntoBST(self, root, val):
        head = root

        # Create a new node
        node = TreeNode(val)

        if not root:
            return node

        while root:
            if root.val < val:
                # Go to the right side of the node
                if root.right:
                    root = root.right

                else:
                    root.right = node
                    break

            elif root.val > val:
                # Go to the left side of the node
                if root.left:
                    root = root.left

                else:
                    root.left = node
                    break

        return head
```
