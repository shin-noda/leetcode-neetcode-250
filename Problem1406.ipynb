```python
class Solution:
    def stoneGameIII(self, stoneValue):
        # Time Complexity: O(N) because we loop through the array size of N exactly once,
        # and the inner loop runs at most 3 times
        # Space Complexity: O(1): we only maintain a list of size 3

        n = len(stoneValue)

        # dp[0], dp[1], dp[2] represent DP[i+1], DP[i+2], DP[i+3] respectively.
        # Initially, beyond the array, the score is 0.
        dp = [0, 0, 0]

        # Iterate backwards from the end of the stoneValue array
        for i in range(n - 1, -1, -1):
            max_relative_score = float('-inf')
            current_stone_sum = 0

            # The player can take 1, 2, or 3 stones
            # This part moves forward, not backwards.
            for k in range(1, 4):
                # This i + k is not an index we're accessing.
                # It's representing the position where the next player starts
                # after taking k stones.
                if i + k <= n:
                    # After taking k stones, the game continues at i + k.
                    current_stone_sum += stoneValue[i + k - 1]

                    # current_stone_sum is what current player gets
                    # dp[k - 1] is the optimal relative score the NEXT player gets
                    score = current_stone_sum - dp[k - 1]
                    max_relative_score = max(max_relative_score, score)

            # Shift our DP states backwards
            # New DP[i] becomes the dp[0] for the next iteration (i - 1)
            # Note: dp[i] = the maximum score difference (current player - other player)
            # starting from index i.
            dp = [max_relative_score, dp[0], dp[1]]

        alice_relative_score = dp[0]

        if alice_relative_score > 0:
            return "Alice"

        elif alice_relative_score < 0:
            return "Bob"

        return "Tie"
```

```python
solution = Solution()
```

```python
stoneValue = [1,2,3,7]

print(solution.stoneGameIII(stoneValue))
```
