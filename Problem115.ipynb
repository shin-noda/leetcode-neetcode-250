```python
from collections import deque

class Solution:
    def numDistinct(self, s, t):
        # Idea:
        # Use KMP to solve this problem.
        # On a second though, no.
        # It doesn't work for a case like s = "asbscs", t = "abc"

        # Maybe 97. Interleaving String could be helpful.

        # I can create s_list = list(s) and eliminate characters
        # which are not in t.

        # I think the constraint is t.

        # Naive approach:
        # Backtracking and if i-th character of s matches with the 0-th
        # character of t, we increaten i to i + 1. Then, compare until
        # it hits the base case that len(matching_count) == len(t) or
        # len(s) - i >= len(t) because you need at least len(t) in len(s)
        # to form the stirng of t.
        #
        # But the problem with this approach is there are a lot of redundancies.

        # Another idea:
        # This is not a feasible because it's computationally slow but
        # for this example: s = "rabbbit", t = "rabbit"
        # we focus t's characters and map idices of s that matches with the
        # char in t.
        # t[0] = 'r'. In s, r[0] = 'r'. So it's (0).
        # t[1] = 'a'. r[1] = 'a'. So, it's (1).
        # t[2] = 'b'. r[2], r[3], and r[4] are 'b'. So, it's (2, 3, 4)
        # and we continue.
        # In the end, we find a combinations that the tuples meet
        # (0) > (1) > (2) ... like strictly following this increasing order
        # and count the valid combinations.
        # It's like a N-Pointer approach.
        #
        # You know what :O ?
        # Maybe this approach works :)
        # Study 18. 4Sum and the extension of NSum
        # but use DP with a huge input like nums.length >= 10k
        # with Nsum.
        #
        # Actually, it doen's work because this NSum doesn't care about the order
        # like 1 + 2 + 3 = 2 + 1 + 3.
        # But for this problem, we must preserve the order of t.

        # Maybe a variation of 0/1 Knapsack problem?
        # For each s[i], we choose either take it or not take it?

        # Maybe not because Knapsack problem doens't care about the order
        # how they pack items.

        # For indecies in s, there are valid indecies that the character of
        # t can jump from left to right.
        # Should we count this configurations?

        # This is a combinatorics problem.

        # Instead of implementing normal N-Pointer, maybe we can
        # use a jumping index pointer like a letter 'b' can only
        # move on letter b's in s.


        # We use N-pointer with jumping indecies approach

        # Check if the length of s is greater than equal to t
        if len(s) < len(t):
            return 0

        # Initialize m[i] where i = 0, 1, .., len(t)
        m = {}
        for i in range(len(t)):
            m[i] = None

        # Record indecies of 's' that matches with t[i].
        ch_map = {}
        for i, ch in enumerate(s):
            if ch not in ch_map:
                ch_map[ch] = []

            ch_map[ch].append(i)

        # O(N^2)
        for i, ch in enumerate(t):
            # Quick exit
            if ch not in ch_map:
                # No subsequences of s which equals to t
                return 0

            # O(N)
            m[i] = deque(ch_map[ch])

        prev_start = m[0][0]
        prev_last = m[len(t) - 1][-1]
        # We shrink list
        for i in range(1, len(t)):
            while m[i] and m[i][0] <= prev_start:
                m[i].popleft()

            while m[len(t) - i - 1] and m[len(t) - i - 1][-1] >= prev_last:
                m[len(t) - i - 1].pop()

            # Another quick exit
            if not m[i] or not m[len(t) - i - 1]:
                return 0

            # Update variables AFTER we confirm that they exist
            prev_start = m[i][0]
            prev_last = m[len(t) - i - 1][-1]

        # Finally, count legal combinations
        # Use backtracking
        # Actually, it's too slow, so we need to find an alternative.
        # Build a 2D dp table
        # Note: this dp table is neither square nor rectagle.
        # dp[i][j] represents the number of valid combinations to form t[:i + 1]
        # when you pick s[m[i][j]]
        dp = []
        for i in range(len(m)):
            dp.append([0] * len(m[i]))

        # Initialize the first row
        for i in range(len(dp[0])):
            dp[0][i] = i + 1

        # Process remaining rows in dp
        for r in range(1, len(dp)):
            for c in range(len(dp[r])):
                counts = 1
                last = dp[r - 1][0]

                while counts < len(m[r - 1]):
                    if m[r - 1][counts] >= m[r][c]:
                        break

                    last = dp[r - 1][counts]
                    counts += 1

                # Update the dp table
                dp[r][c] = last

                if c > 0:
                    dp[r][c] += dp[r][c - 1]

        return dp[-1][-1]
```
