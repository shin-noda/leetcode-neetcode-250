```python
class Solution:
    def numRescueBoats(self, people, limit):
        # Sort the array
        people.sort()

        n = len(people)

        boats = 0
        left = 0
        right = n - 1

        while left <= right:
            if people[left] + people[right] <= limit:
                # We can carry both people.
                # Otherwise, we just carry the right one because since the list is sorted, we
                # cannot carry the right person with someone later either.
                left += 1

            right -= 1
            boats += 1

        return boats
```

```python
solution = Solution()
```

```python
people = [1,2]
limit = 3

print(solution.numRescueBoats(people, limit))
```
