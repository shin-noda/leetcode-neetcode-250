```python
class Solution:
    def largestRectangleArea(self, heights):
        # Brute force approach: for each index, check the
        # leftmost and rightmost indecies either the end of list
        # or its value has zero and calculate min(val) * width

        # Another idea: Create a set and keep track of rectangles whose heights are
        # 1, 2, ..., max(heights) and find the maximum among those rectangles

        # Use Monotonic Stack with tuple packing
        # Mnonotic Stack with increasing order and the width
        # contains the "history" of width

        n = len(heights)
        max_area = 0

        # We store (height, width) for each element
        stack = []

        for i in range(n):
            height = heights[i]
            width = 0
            while stack and stack[-1][0] > height:
                h, w = stack.pop()

                # Calculate the max_area
                max_area = max(max_area, h * (w + width))

                width += w

            stack.append((height, width + 1))

        # We calculate the max_area remaining in the Monotonic stack
        width = 0
        for i in range(len(stack) - 1, -1, -1):
            h, w = stack.pop()

            max_area = max(max_area, h * (w + width))

            width += w

        return max_area
```

```python
solution = Solution()
```
