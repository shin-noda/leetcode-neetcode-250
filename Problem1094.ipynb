```python
import heapq

class Solution:
    def carPooling(self, trips, capacity):
        # trips[i] = [numPassengers_i, from_i, to_i]

        # We sort trips with 'from', and then, 'to'.
        trips.sort(key=lambda x: (x[1], x[2]), reverse=True)
        min_heap = []
        curr = 0

        while trips or min_heap:
            pickup = float('inf')
            dropoff = float('inf')

            if trips:
                pickup = trips[-1][1]

            if min_heap:
                dropoff = min_heap[0][0]

            # Advance the car to the next dropoff or picking up point.
            # We want to dropoff passgengers first if there are any
            # to increase the capacity
            curr = min(pickup, dropoff)

            # For each loop, we do either dropping off or picking up
            # If dropping off and picking up happen at the same point, we
            # prioritize dropping off to increase the capacity
            if min_heap and curr == dropoff:
                # Drop off passengers if any
                while min_heap and curr == min_heap[0][0]:
                    t, num = heapq.heappop(min_heap)

                    # Increase the capacity
                    capacity += num

            # Pick up passangers
            else:
                while trips and curr == trips[-1][1]:
                    num, f, t = trips.pop()

                    # Reduce the capacity
                    capacity -= num

                    # Check if it's out of capacity
                    if capacity < 0:
                        return False

                    # We store by the 'to'. 'from' is irrelevant now because we've already picked the up.
                    # 'numPassengers' are also irrelevant because we drop everyone at the same time at
                    # the destination and order of nums doesn't matter.
                    heapq.heappush(min_heap, (t, num))

        return True
```
