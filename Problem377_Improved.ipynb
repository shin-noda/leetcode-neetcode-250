```python
class Solution:
    def combinationSum4(self, nums, target):
        # dp[i] stores the number of combinations to reach target i
        dp = [0] * (target + 1)

        # Base Case: 1 way to make 0 (by choosing nothing)
        dp[0] = 1

        # Early break optimization
        nums.sort()

        for i in range(1, target + 1):
            for num in nums:
                if num > i:
                    # Since nums is sorted, no larger numbers will fit either
                    break

                dp[i] += dp[i - num]

        return dp[target]
```

```python
solution = Solution()
```

```python
nums = [1,2,3]
target = 4

print(solution.combinationSum4(nums, target))
```
