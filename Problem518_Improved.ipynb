```python
class Solution:
    def change(self, amount, coins):
        # Idea:
        # This is a Knapsack Problem but unbounded because
        # use can reuse the same coin multiple times.

        # - Knapsack Capacity: 'amount'
        # - Items: 'coins'. Each coin can either be included in the amount or left out
        # - Objective: Put collect coins to sum the exact same value as 'amount'.

        # Sort coins in ascending (increasing) order.
        # c0, c1, c2, ... cn
        # We are gonna find all possible combinations the meets:
        # a0*c0 + a1*c1 + a2*c2 + ... + an*cn = amount

        # The brute force way:
        # Use backtracking with index and the base case would be
        # the sum is greater than or equal to 'amount'.
        # The short circuit would be when ak > 'amount', we terminate it.
        #
        # The propble of this approach is that you do "unchoose", so
        # there are gonna be redundant combinations.
        #
        # Instead, we're gonna use DP.
        # The dp table represents:
        # the numbers of way to sum up to that amount.
        # e.g. amount = 5, coins = [1,2,5]
        # dp[0] is 1 by not choosing anything from coins.
        # dp[1] is 1: [1]
        # dp[2] is 2: [1,1] or [2]
        # dp[3] is 2: [1,1,1] or [1,2]
        # dp[4] is 3: [1,1,1,1], [1,1,2] or [2,2]
        # dp[5] is 4: [1,1,1,1,1], [1,1,1,2], [1,2,2] or [5]

        # Is this a staricase problem?
        # To get to the fifth floor:
        # You go there from:
        # - the 0th floor with 5 steps, (dp[0] = 1)
        # - the 3rd floor with 2 steps, or (dp[3] = 2)
        # - the 4th floor with 1 step. (dp[1] = 1)
        #
        # In general, to get dp[i]:
        # dp[i] = dp[i - c0] + dp[i - c1] + ... + dp[i - cn]
        # where i - cn >= 0.
        #
        # There is a problem with this approach because there are duplicates.
        # For example, for dp[3]:
        # To get to the 3rd floor, you can either take 1 step from the 2nd floor or
        # 2 steps from the 1st floor.
        # So dp[3] = dp[1] + dp[2] = 3.
        # Which is {(1,1,1)} + {(1,1,1) + (2,1)}.
        # But as we can see, (1,1,1) is a duplicate.
        # We have to account for this duplicate.
        #
        # To fix this problem, we may need to iterate something backwards.
        # What would that be...?
        # Maybe we can sort coins backwards and skip steps with coin steps
        # For example, if coin = 5, we record dp[5], dp[10], etc.
        # First, we do dp[5] += 1.
        # And then, dp[10] += dp[5], dp[15] += dp[10], etc.
        # Since we iterate coins from a large number, there's not gonna be any duplicates.
        #
        # This problem has a problem because it undercounts combinations.
        #
        # I think I should use the original idea:
        # dp[i] = dp[i - c0] + dp[i - c1] + ... + dp[i - cn]
        # but tweak it so that we can avoid overcounting.

        # Note: we don't need to sort coins.
        # Because the outer loop ensures we process one coin at a time, the
        # actual numerical order of the coins doesn't affect the combination lock-in.
        # This is a combination problem, so order doesn't change the outcome.

        # Note: it's (amount + 1), not 'amount' because we make a dp table from 0 to 'amount'.
        dp = [0] * (amount + 1)

        # We can make 0 with not choosing any coin
        dp[0] = 1

        for coin in coins:
            for i in range(coin, amount + 1):
                dp[i] += dp[i - coin]

        return dp[amount]
```
