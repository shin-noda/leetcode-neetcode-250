```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
class Solution:
    def insertGreatestCommonDivisors(self, head):
        # Time Complexity: O(N log(min(A, B)))

        # Helper function
        def gcd(num1, num2):
            while num2 > 0:
                num1, num2 = num2, num1 % num2

            return num1

        # A sentinel node
        dummy = ListNode()
        dummy.next = head

        while head.next:
            # Get two values
            num1 = head.val
            num2 = head.next.val

            # Have a copy of the next node
            next_node = head.next

            # Create a new gcd node
            gcd_node = ListNode(gcd(num1, num2))

            # Attach the new node between the current and next nodes
            head.next = gcd_node
            gcd_node.next = next_node

            # Proceed to the next node
            head = next_node

        return dummy.next
```
