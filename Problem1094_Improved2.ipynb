```python
import heapq

class Solution:
    def carPooling(self, trips, capacity):
        # Use Difference Array

        # Create a timeline map of all possible locations (0 to 1000)
        # We need 1001 slots to safely index up to 1000
        delta = [0] * 1001

        # Record only the change in passenger count at boundaries
        for num, f, t in trips:
            # Passengers board here (+net load)
            delta[f] += num

            # Passengers leave here (-net load)
            delta[t] -= num

        # Roll through the timeline and calculate the running sum of passengers
        curr = 0
        for change in delta:
            curr += change
            if curr > capacity:
                return False

        return True
```
