```python
import heapq

class Solution:
    def topKFrequent(self, nums, k):
        # First, count the occurrences of numbers in the array
        m = {}

        # O(N)
        for num in nums:
            m[num] = m.get(num, 0) + 1

        n = len(m)
        pairs = []

        # O(N)
        for key in m:
            ocurr = m[key]

            # The order is important. We want the occurances first
            pairs.append((ocurr, key))

        # Create a min heap: O(k)
        heap = pairs[:k]
        heapq.heapify(heap)

        # O((N - k)log k)
        for i in range(k, n):
            if heap[0][0] < pairs[i][0]:
                heapq.heapreplace(heap, pairs[i])

        results = []
        for occur, key in heap:
            results.append(key)

        return results
```

```python
solution = Solution()
```

```python
nums = [1,1,1,2,2,3]
k = 2

print(solution.topKFrequent(nums, k))
```
