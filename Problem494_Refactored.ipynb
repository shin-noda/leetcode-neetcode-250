```python
class Solution:
    def findTargetSumWays(self, nums, target):
        # Initialize dp[0] with 1
        # We update dp in for-loop for each iteration, so
        # If 'target' is 0 and the final sums do not end up with 0, it
        # won't be an issue
        dp = {0: 1}

        for num in nums:
            next_dp = {}

            for s, count in dp.items():
                next_dp[s + num] = next_dp.get(s + num, 0) + count
                next_dp[s - num] = next_dp.get(s - num, 0) + count

            # Update dp
            dp = next_dp

        # If the final sum does not end up with 'target', it should return 0.
        return dp.get(target, 0)
```

```python
solution = Solution()
```

```python
nums = [1,1,1,1,1]
target = 3

print(solution.findTargetSumWays(nums, target))
```

```python
nums = [1]
target = 1

print(solution.findTargetSumWays(nums, target))
```

```python
nums = [1,1]
target = 0

# It should return 2
print(solution.findTargetSumWays(nums, target))
```

```python
nums = [0,0,0]
target = 0

# It should return 8 (2^3 = 8)
print(solution.findTargetSumWays(nums, target))
```

```python
nums = [1]
target = 5

print(solution.findTargetSumWays(nums, target))
```
