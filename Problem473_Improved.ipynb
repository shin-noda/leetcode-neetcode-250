```python
class Solution:
    def makesquare(self, matchsticks):
        # Quick check: the sum of matchsticks should be a multiple of 4.
        # Also, there should be at least 4 elements in them because we cannot' break them.

        # Another quick check
        # The bottlenect of the lengh of the edge is
        # max_val = max(matchsticks)
        # because we cannot break match sticks, so if max_val > sum(matchsticks) // 4, then
        # it's impossible to form a square

        matchsticks.sort(reverse=True)
        sum_val = sum(matchsticks)
        n = len(matchsticks)
        edge = sum_val // 4

        if not matchsticks or sum_val % 4 != 0 or n < 4 or matchsticks[0] > edge:
            return False

        self.is_achievable = False
        self.n = n
        self.edge = edge
        self.edges = [0, 0, 0, 0]


        def backtrack(count):
            n = self.n
            edge_len = self.edge

            # Base Case:
            if self.is_achievable or count >= n:
                self.is_achievable = True
                return

            for i in range(len(self.edges)):
                # Choose
                m = matchsticks[count]
                self.edges[i] += m

                # Explore
                if self.edges[i] <= edge_len:
                    backtrack(count + 1)

                # Backtrack
                self.edges[i] -= m

                # The pruning
                # If this edge went back to 0, it means putting 'm' here failed.
                # Because all subsequent empty edges are identical, 'm' will fail there too.
                # Thus, we break early to prevent redundant, identical deep searches
                if self.edges[i] == 0:
                    break

        backtrack(0)

        return self.is_achievable
```
