```python
import heapq

class Solution:
    def networkDelayTime(self, times, n, k):
        # Build adjacency list using a standard list of lists
        # Index represents the source node. Size is n + 1
        adj = [[] for _ in range(n + 1)]
        for u, v, w in times:
            adj[u].append((v, w))

        # Initialize Dijkstra structures
        min_heap = [(0, k)]
        visited = set()

        while min_heap:
            t, curr = heapq.heappop(min_heap)

            # By Dijkstra's Greedy nature, it's guaranteed that
            # visited nodes are the shortest paths
            if curr in visited:
                continue

            visited.add(curr)

            # Also Dijkstra's Greedy nature, this t is guaranteed
            # to the the minimum time
            if len(visited) == n:
                return t

            # Explore neighbours safely
            # Since adj[curr] always exists as an empty list or filled list, we
            # don't have to check "if curr in adj" anymore
            for neighbour, weight in adj[curr]:
                if neighbour not in visited:
                    heapq.heappush(min_heap, (t + weight, neighbour))

        return -1
```
