```python
class Solution:
    def getSum(self, a, b):
        # 32-bit mask to bound Python's infinite precision
        mask = 0xFFFFFFFF

        while b != 0:
            # '^' finds the sum without carries, bounded to 32 bits
            # '&' finds the carries, << 1 shifts them left, bounded to 32 bits
            a, b = (a ^ b) & mask, ((a & b) << 1) & mask

        # If 'a' is a negative number in 32-bit Two's Complement,
        # its 31st bit (the sign bit) will be 1.
        # 0xFFFFFFF is the largest positive 32-bit signed integer
        if a > 0x7FFFFFFF:
            # Convert the 32-bit unsigned representation back to Python's signed form
            return ~(a ^ mask)

        return a
```

```python
solution = Solution()
```

```python
a = 1
b = 2

print(solution.getSum(a, b))
```
