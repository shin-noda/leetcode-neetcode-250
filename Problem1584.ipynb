```python
import heapq

class Solution:
    def minCostConnectPoints(self, points):
        # Use Dijkstra's Algorithm
        # On a second though, nope.

        # This problem visits all nodes exactly once -> Hamiltonia Path.
        # Nope :)

        # This is a minimum spanning tree problem, so use Prim's.
        # Time Complexity: O((V + E)logV) where V is the number of vertices and
        # E is the number of edges

        # Check if there is only one point
        if len(points) == 1:
            return 0

        adj = [[float('inf')] * len(points) for _ in range(len(points))]
        for i in range(len(points)):
            for j in range(i + 1, len(points)):
                # The Manhattan Distance
                md = abs(points[i][0] - points[j][0]) + abs(points[i][1] - points[j][1])

                adj[i][j] = md
                adj[j][i] = md

        total_cost = 0
        visited = set()
        # (weight, current_node)
        min_heap = [(0, 0)]

        while min_heap:
            w, curr = heapq.heappop(min_heap)

            if curr in visited:
                continue

            # Mark the current node
            visited.add(curr)

            # Add the cost
            total_cost += w

            for i in range(len(adj)):
                if i != curr and i not in visited:
                    heapq.heappush(min_heap, (adj[curr][i], i))

        return total_cost
```
