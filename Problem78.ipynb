```python
class Solution:
    def subsets(self, nums):
        self.tmp = []
        self.results = set(())

        def backtrack(index):
            # Base case
            if index >= len(nums):
                return

            for i in range(index, len(nums)):
                # Choose
                self.tmp.append(nums[i])
                self.results.add(tuple(self.tmp))

                # Explore
                backtrack(i + 1)

                # Unchoose
                self.tmp.pop()

        backtrack(0)

        # Convert a set to a list
        answers = [list(x) for x in self.results]

        # Add an empty set
        answers.append([])

        return answers
```
