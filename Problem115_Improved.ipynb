```python
class Solution:
    def numDistinct(self, s, t):
        m, n = len(t), len(s)
        if n < m:
            return 0

        # dp[i][j] -> t[:i] and s[:j]
        # dp[i][j]: number of ways to form the first i characters of t
        # using only the first j characters of s.
        # - t[:i] is our target prefix
        # - s[:j] is our available prefix

        # Key question
        # - What can we do with s[j - 1] ?
        # At position (i, j), we're looking at:
        # - target character: t[i - 1]
        # - source character: s[j - 1]
        # There are two possibilities for s[j - 1]
        #
        # 1. We do not use s[j - 1]
        # -> Nothing changes execpt we ignore it.
        # So, all ways to form t[:i] using s[:j - 1] that is:
        # - dp[i][j - 1]
        #
        # 2. You do use s[j - 1] (only if it matches)
        # -> We can only use it if:
        # - t[i - 1] == s[j - 1]
        # If we use it, then:
        # that character must be the last character used to form t[:i]
        # so before using it, you must have already formed:
        # - t[:i - 1]
        # - using s[:j - 1]
        # So:
        # - dp[i - 1][j - 1]
        #
        # Why do we add them?
        # Because these two cases are:
        # - disjoint (they use different decisions about s[j - 1])
        # - cover all possibilities
        # So:
        # dp[i][j] = (skip s[j - 1]) + (use s[j - 1 if possible])

        dp = [[0] * (n + 1) for _ in range(m + 1)]

        # Base case: An empty string t is a subsequence of any prefix of s exactly 1 way
        for j in range(n + 1):
            dp[0][j] = 1

        for i in range(1, m + 1):
            for j in range(1, n + 1):
                # We can always choose to skip s[j - 1]
                dp[i][j] = dp[i][j - 1]

                # If characters match, we can also choose to take s[j - 1]
                if t[i - 1] == s[j - 1]:
                    dp[i][j] += dp[i - 1][j - 1]

        return dp[m][n]
```

```python
solution = Solution()
```
