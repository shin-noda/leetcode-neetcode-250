```python
class Solution:
    def numDistinct(self, s, t):
        m, n = len(t), len(s)
        if n < m:
            return 0

        # dp[i][j] -> t[:i] and s[:j]
        dp = [[0] * (n + 1) for _ in range(m + 1)]

        # Base case: An empty string t is a subsequence of any prefix of s exactly 1 way
        for j in range(n + 1):
            dp[0][j] = 1

        for i in range(1, m + 1):
            for j in range(1, n + 1):
                # We can always choose to skip s[j - 1]
                dp[i][j] = dp[i][j - 1]

                # If characters match, we can also choose to take s[j - 1]
                if t[i - 1] == s[j - 1]:
                    dp[i][j] += dp[i - 1][j - 1]

        return dp[m][n]
```

```python
solution = Solution()
```

```python
s = "rabbbit"
t = "rabbit"

print(solution.numDistinct(s, t))
```
