```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
from collections import deque

class Solution:
    def isSubtree(self, root, subRoot):
        self.is_same = False
        self.left_depth = 0
        self.right_depth = 0

        def dfs(node, isComparing):
            if not node or self.is_same:
                return 0

            left = dfs(node.left, isComparing)
            right = dfs(node.right, isComparing)

            if (
                isComparing and
                (left == self.left_depth and right == self.right_depth) and
                compareTwoNodes(node, subRoot)
            ):
                self.is_same = True

            return 1 + max(left, right)


        def compareTwoNodes(node1, node2):
            dq1 = deque([node1])
            dq2 = deque([node2])

            while dq1 and dq2:
                n1 = len(dq1)
                n2 = len(dq2)

                if n1 != n2:
                    return False

                for i in range(n1):
                    item1 = dq1.popleft()
                    item2 = dq2.popleft()

                    if item1.val != item2.val:
                        return False

                    if item1.left:
                        dq1.append(item1.left)

                    if item1.right:
                        dq1.append(item1.right)

                    if item2.left:
                        dq2.append(item2.left)

                    if item2.right:
                        dq2.append(item2.right)

            return len(dq1) == len(dq2)

        self.left_depth = dfs(subRoot.left, False)
        self.right_depth = dfs(subRoot.right, False)

        dfs(root, True)

        return self.is_same
```
