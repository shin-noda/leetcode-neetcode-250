```python
class MaxHeap:
    def __init__(self):
        self.heap = []


    def heapify_down(self, idx, n):
        while True:
            largest = idx
            left = 2 * idx + 1
            right = 2 * idx + 2

            if left < n and self.heap[largest] < self.heap[left]:
                largest = left

            if right < n and self.heap[largest] < self.heap[right]:
                largest = right

            if largest == idx:
                break

            else:
                self.heap[largest], self.heap[idx] = self.heap[idx], self.heap[largest]
                idx = largest


class Solution:
    def sortArray(self, nums):
        heap = MaxHeap()
        n = len(nums)

        # We sort it in-place
        heap.heap = nums

        # Build a heap
        for i in range((n // 2) - 1, -1, -1):
            heap.heapify_down(i, n)

        # Heap Sort
        for i in range(n - 1, -1, -1):
            # Swap the first element (max) and the last element
            heap.heap[0], heap.heap[i] = heap.heap[i], heap.heap[0]

            # Sift down (Reduce the array size by 1)
            heap.heapify_down(0, i)

        return nums
```

```python
solution = Solution()
```

```python
nums = [5,2,3,1]

print(solution.sortArray(nums))
```
