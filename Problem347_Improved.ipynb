```python
import heapq

class Solution:
    def topKFrequent(self, nums, k):
        # Use Bucket Sort to achieve O(N) time complexity

        # Count frequencies: O(N)
        count = {}
        for num in nums:
            count[num] = count.get(num, 0) + 1

        # Initialize buckets: O(N)
        # index is frequency, value is list of numbers with that frequency
        buckets = [[] for _ in range(len(nums) + 1)]

        # Fill buckets: O(N)
        for n, c in count.items():
            buckets[c].append(n)

        # Collect the top k: O(N)
        results = []

        # Iterate backwards from the highest possible frequency
        for i in range(len(buckets) - 1, 0, -1):
            for n in buckets[i]:
                results.append(n)

                if len(results) == k:
                    return results
```

```python
solution = Solution()
```

```python
nums = [1,1,1,2,2,3]
k = 2

print(solution.topKFrequent(nums, k))
```
