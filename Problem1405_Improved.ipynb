```python
class Solution:
    def longestDiverseString(self, a, b, c):
        # Keep track of counts dynamically in a simple list
        # Note: Each entry means [count_remaining, character], so
        # every time we use characters, we reduce its remaining count.
        counts = [[a, 'a'], [b, 'b'], [c, 'c']]
        results = []

        while True:
            # Sort descending by count so the largest is alwasy at index 0
            counts.sort(key=lambda x: x[0], reverse=True)

            # If the largest available count is 0, we are completely done
            if counts[0][0] == 0:
                break

            # Check if using the largest element would violate the "no triple" rule
            # Note: we have len(results) >= 2 because we want to inspect
            # results[-1] and results[-2], so make sure they exist.
            if len(results) >= 2 and results[-1] == counts[0][1] and results[-2] == counts[0][1]:
                # We can't use the largest, so check if a second element exists
                if counts[1][0] == 0:
                    # No second element available to break the streak, so we must stop
                    break

                # Pick the second largest instead
                results.append(counts[1][1])
                counts[1][0] -= 1

            else:
                # Safe to use the largest element
                results.append(counts[0][1])
                counts[0][0] -= 1

        return "".join(results)
```
