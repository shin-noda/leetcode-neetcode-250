```python
class Solution:
    def carFleet(self, target, position, speed):
        # Check if there is only one car
        if len(position) <= 1:
            return 1

        fleets = 1
        n = len(position)

        # Tuple packing
        tmp = []
        for i in range(n):
            tmp.append((position[i], speed[i]))

        # Sort by position, decreasing order
        cars = sorted(tmp, key=lambda v: v[0], reverse=True)

        # We only care about the fleet in front of the current car becasue cars
        # are sorted by position, so if the current car cannot catch up with the front fleet, it
        # won't join any fleet unless car behind join the current one
        fp, fs = cars[0]
        for i in range(1, n):
            p, s = cars[i]

            # The current car won't join the fleet in front of it, so
            # the current car becomes the next bottle neck
            if s <= fs:
                fp = p
                fs = s

                # There's gonna be a new fleet
                fleets += 1

            else:
                # Note: since s != fs, we don't have to worry about division by 0
                x = (fp - p) / (s - fs)

                # This also works: meeting_point = fp + fs * x
                meeting_point = p + s * x

                if meeting_point > target:
                    # The current car is a new bottle neck
                    fp = p
                    fs = s

                    fleets += 1


        return fleets
```

```python
solution = Solution()
```
