```python
class Solution:
    def findDuplicate(self, nums):
        # Since we need to solve this problem in O(1) space complexity, we
        # use nums, the ipnut value and Cyclic Sort

        n = len(nums)
        hold = 0

        for i in range(n):
            if nums[i] == i + 1:
                # The number is placed in the right location
                continue

            else:
                hold = nums[i]

                # Mark as -1
                nums[i] = -1

                while hold != -1:
                    if nums[hold - 1] == hold:
                        return hold

                    # Have a backup
                    tmp = nums[hold - 1]

                    # Assign the right value to the right place
                    nums[hold - 1] = hold

                    # Update the holding value
                    hold = tmp
```

```python
solution = Solution()
```

```python
nums = [1,3,4,2,2]

print(solution.findDuplicate(nums))
```
