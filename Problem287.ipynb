```python
class Solution:
    def findDuplicate(self, nums):
        # Finding the intersection point in the cycle
        # We treat the array as a linked list: index -> value
        slow = nums[0]
        fast = nums[nums[0]]

        while slow != fast:
            slow = nums[slow]
            fast = nums[nums[fast]]

        # Finding the entrance to the cyle
        # Reset one pointer to the start
        slow = 0
        while slow != fast:
            # Note: During the fist jump, slow is at index 0,
            # so we move to nums[0]
            if slow == 0:
                slow = nums[0]

            else:
                slow = nums[slow]

            fast = nums[fast]

        return slow
```

```python
solution = Solution()
```

```python
nums = [1,3,4,2,2]

print(solution.findDuplicate(nums))
```
