```python
class Solution:
    def wordBreak(self, s, wordDict):
        # O(1) lookups
        wordSet = set(wordDict)
        combo = []
        results = []

        max_len = max(len(w) for w in wordDict)


        def backtrack(index):
            if index == len(s):
                results.append(" ".join(combo))
                return

            # Loop dynamically up to max_len
            for end in range(index + 1, min(index + max_len + 1, len(s) + 1)):
                word = s[index:end]
                if word in wordSet:
                    # Choose
                    combo.append(word)

                    # Explore
                    backtrack(end)

                    # Unchoose
                    combo.pop()


        backtrack(0)

        return results
```
