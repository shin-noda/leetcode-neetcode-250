```python
from collections import deque

class Solution:
    def findCheapestPrice(self, n, flights, src, dst, k):
        # Use Dijkstra's Algorithm

        adj = [[] for _ in range(n)]
        costs = [float('inf')] * n

        # Mark the staring point as 0
        costs[src] = 0

        # From, To, Price
        for f, t, p in flights:
            adj[f].append((t, p))

        # (total_cost, from, stops)
        queue = deque([(0, src, 0)])

        while queue:
            cost, curr, stops = queue.popleft()

            n = len(adj[curr])
            for i in range(n):
                next_stop, next_cost = adj[curr][i]

                total_cost = cost + next_cost
                # Note: the second condition is stops <= k, not stops + 1 <= k
                if total_cost < costs[next_stop] and stops <= k:
                    costs[next_stop] = total_cost
                    queue.append((total_cost, next_stop, stops + 1))

        if costs[dst] != float('inf'):
            return costs[dst]

        return -1
```
