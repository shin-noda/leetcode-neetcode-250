```python
class Solution:
    def partition(self, s):
        n = len(s)

        # Build the 2D DP table for O(1) palindrome lookups
        # dp[i][j] will be True if s[i:j + 1] is a palindrome
        dp = [[False] * n for _ in range(n)]

        # Populate the table from the bottom up
        for i in range(n - 1, -1, -1):
            for j in range(i, n):
                if s[i] == s[j]:
                    # Two cases where it's a palindrome:
                    # - The characters are adjacent or identical (j - i <= 2)
                    # - The inner substring is also a palindrome (dp[i + 1][j - 1])
                    if j - i <= 2 or dp[i + 1][j - 1]:
                        dp[i][j] = True

        results = []
        path = []

        # Pure Backtracking using the DP table
        def backtrack(start):
            if start >= n:
                results.append(list(path))

                return

            for i in range(start, n):
                # Instead of scanning characters, do a O(1) check
                if dp[start][i]:
                    # Choose
                    path.append(s[start:i + 1])

                    # Explore
                    backtrack(i + 1)

                    # Unchoose (Zero-allocation cleanup)
                    path.pop()


        backtrack(0)

        return results
```

```python
solution = Solution()
```

```python
s = "aab"

print(solution.partition(s))
```
