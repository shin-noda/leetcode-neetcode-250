```python
class Solution:
    def lemonadeChange(self, bills):
        n = len(bills)

        # Each represents the number of $5 and $10 bills.
        # Initially, we don't have any change
        # Note: we don't keep track of $20 bills because we don't need it.
        b_5 = 0
        b_10 = 0

        # Quick check: if the first customer pays with a $10 or $20 bill, we don't have change.
        if not bills or bills[0] != 5:
            return False

        # The first customer paid with a $5 bill
        b_5 += 1
        for i in range(1, n):
            bill = bills[i]

            # No change needed
            if bill == 5:
                b_5 += 1

            elif bill == 10:
                if b_5 <= 0:
                    return False

                b_5 -= 1
                b_10 += 1

            else:
                # We always give customers a $10 first because
                # it has more flexibility
                owed = 15

                if b_10 > 0:
                    b_10 -= 1
                    owed -= 10

                if b_5 >= owed // 5:
                    b_5 -= owed // 5
                    owed = 0

                if owed != 0:
                    return False

        return True
```

```python
solution = Solution()
```

```python
bills = [5,5,5,10,20]

print(solution.lemonadeChange(bills))
```

```python
bills = [5,5,10,10,20]

print(solution.lemonadeChange(bills))
```
