```python
import heapq

class Solution:
    def minimumEffortPath(self, heights):
        # Use Dijkstra's Algorithm

        rows, cols = len(heights), len(heights[0])

        # Matrix to store the minimum effort to reach each cell
        efforts = [[float('inf')] * cols for _ in range(rows)]
        efforts[0][0] = 0

        # Min-heap stores tuples of: (effort, row, col)
        min_heap = [(0, 0, 0)]
        directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

        while min_heap:
            current_effort, r, c = heapq.heappop(min_heap)

            # If we reached the destination, this is the minimum effort path
            if r == rows - 1 and c == cols - 1:
                return current_effort

            # If we found a worse path than already recoded, skip it
            if current_effort > efforts[r][c]:
                continue

            for dr, dc in directions:
                nr, nc = r + dr, c + dc

                # Check bounds
                if 0 <= nr < rows and 0 <= nc < cols:
                    new_effort = max(current_effort, abs(heights[r][c] - heights[nr][nc]))

                    # If this path offers a lower effort to the neighbour, update and push it.
                    if new_effort < efforts[nr][nc]:
                        efforts[nr][nc] = new_effort
                        heapq.heappush(min_heap, (new_effort, nr, nc))

        return 0
```
