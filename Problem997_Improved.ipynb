```python
class UnionFind:
    def __init__(self, size):
        self.parent = list(range(size))

class Solution:
    def findJudge(self, n, trust):
        # Idea: in-degree/out-degree

        # Array to store the net trust score: (indegree - outdegree)
        # Size is n + 1 so we can use 1-based indexing directly
        trust_scores = [0] * (n + 1)

        for src, dst in trust:
            # src trusts someone, decrease score
            trust_scores[src] -= 1

            # dst is trusted, increase score
            trust_scores[dst] += 1

        # The judge must have a net score of exactly n - 1
        for i in range(1, n + 1):
            if trust_scores[i] == n - 1:
                return i

        return -1
```
