```python
class Solution:
    def checkInclusion(self, s1, s2):
        # Check if the length of s1 is less than its s2
        if len(s1) > len(s2):
            return False

        s1_freq = {}
        s2_freq = {}

        # The size of window is fixed and it's len(s1)
        # Initialize two dictionaries
        for i in range(len(s1)):
            ch1 = s1[i]
            ch2 = s2[i]

            s1_freq[ch1] = s1_freq.get(ch1, 0) + 1
            s2_freq[ch2] = s2_freq.get(ch2, 0) + 1

        if s1_freq == s2_freq:
            return True

        for i in range(len(s1), len(s2)):
            prev = s2[i - len(s1)]
            curr = s2[i]

            s2_freq[curr] = s2_freq.get(curr, 0) + 1
            s2_freq[prev] = s2_freq[prev] - 1

            # We remove the prev key if it's 0
            if s2_freq[prev] == 0:
                s2_freq.pop(prev)

            # Check if s1's permutations is the substring of s2
            if s1_freq == s2_freq:
                return True

        return False
```

```python
solution = Solution()
```

```python
s1 = "ab"
s2 = "eidbaooo"

print(solution.checkInclusion(s1, s2))
```
