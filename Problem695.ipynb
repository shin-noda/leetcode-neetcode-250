```python
class Solution:
    def maxAreaOfIsland(self, grid):
        max_area = 0

        def dfs(r, c):
            # Basecase
            if (
                not (0 <= r <= len(grid) - 1) or
                not (0 <= c <= len(grid[r]) - 1) or
                grid[r][c] == 0
            ):
                return 0

            # Mark the island
            grid[r][c] = 0

            return (
                1 +
                dfs(r - 1, c) +
                dfs(r + 1, c) +
                dfs(r, c - 1) +
                dfs(r, c + 1)
            )

        for i in range(len(grid)):
            for j in range(len(grid[i])):
                max_area = max(max_area, dfs(i, j))

        return max_area
```
