```python
class Solution:
    def pacificAtlantic(self, heights):
        if not heights or not heights[0]:
            return []

        R, C = len(heights), len(heights[0])
        p_vis = set()
        a_vis = set()

        def dfs(r, c, vis, prev_h):
            # Base Cases
            if (
                r < 0 or r >= R or
                c < 0 or c >= C or
                (r, c) in vis or
                heights[r][c] < prev_h
            ):
                return

            vis.add((r, c))

            # Standard 4-directional crawl
            dfs(r + 1, c, vis, heights[r][c])
            dfs(r - 1, c, vis, heights[r][c])
            dfs(r, c + 1, vis, heights[r][c])
            dfs(r, c - 1, vis, heights[r][c])


        # Start from Left/Right edges
        for r in range(R):
            # Left edge -> Pacific
            dfs(r, 0, p_vis, heights[r][0])

            # Right edge -> Atlantic
            dfs(r, C - 1, a_vis, heights[r][C - 1])

        # Start from Top/Bottom edges
        for c in range(C):
            # Top edge -> Pacific
            dfs(0, c, p_vis, heights[0][c])

            # Bottom edge -> Atlantic
            dfs(R - 1, c, a_vis, heights[R - 1][c])

        # The intersection is our result
        o_vis = p_vis & a_vis
        return [list(cell) for cell in o_vis]
```
