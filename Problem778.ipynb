```python
from collections import deque

class Solution:
    def swimInWater(self, grid):
        # We use Binary Search + BFS

        def can_reachable(water_level):
            rows, cols = len(grid), len(grid[0])

            visited = [[False] * cols for _ in range(rows)]

            # Check if the starting cell has enough water level
            if grid[0][0] > water_level:
                return False

            # Mark the top left cell as visited
            visited[0][0] = True

            # Up, Down, Left and Right
            directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

            # (row, column)
            queue = deque([(0, 0)])
            while queue:
                r, c = queue.popleft()

                if r == rows - 1 and c == cols - 1:
                    # It's accessible
                    return True

                for dr, dc in directions:
                    nr, nc = r + dr, c + dc

                    if (
                        0 <= nr < rows and
                        0 <= nc < cols and
                        not visited[nr][nc] and
                        grid[nr][nc] <= water_level
                    ):
                        queue.append((nr, nc))

                        # Mark as visited
                        visited[nr][nc] = True

            return False


        min_time = 0
        left = 0
        right = len(grid) * len(grid[0]) - 1

        while left <= right:
            mid = (left + right) // 2

            if can_reachable(mid):
                min_time = mid
                right = mid - 1

            else:
                left = mid + 1

        return min_time
```
