```python
class Solution:
    def findMinHeightTrees(self, n, edges):
        # Idea:
        # We build a graph and then, remove leaf nodes one by one.
        # The last remaining either one or two nodes are the answer.
        # The answer is either [n1] or [n1, n2].

        # Check if number of nodes are less than or equal to 2
        if n <= 2:
            return [i for i in range(n)]

        adj = {}
        for i, j in edges:
            if i not in adj:
                adj[i] = set()

            if j not in adj:
                adj[j] = set()

            adj[i].add(j)
            adj[j].add(i)

        leaves = set()
        parents = set()
        for key in adj:
            nodes = adj[key]
            if len(nodes) == 1:
                leaves.add(key)

            else:
                parents.add(key)

        # We peel nodes like onion until remaining parents is either 1 or 2 numbers.
        while len(parents) > 2:
            n = len(leaves)
            tmp = set()

            for _ in range(n):
                leaf = leaves.pop()
                next = adj[leaf].pop()

                # Delete the leaf key in adj
                adj.pop(leaf)

                # Remove 'leaf' in the 'next' key
                adj[next].remove(leaf)

                # Remove leaf from parents if it exists
                if leaf in parents:
                    parents.remove(leaf)

                # Check if adj[next] is a new leaf node now
                if len(adj[next]) == 1:
                    tmp.add(next)

            leaves = tmp

        return list(parents)
```
