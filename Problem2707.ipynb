```python
class TrieNode:
    def __init__(self, val=None):
        self.val = val
        self.nodes = [None] * 26
        self.is_end = False


class Trie:
    def __init__(self):
        self.head = TrieNode()


    def _ch_to_idx(self, ch):
        return ord(ch) - ord('a')


    def insert(self, chars):
        head = self.head

        for ch in chars:
            idx = self._ch_to_idx(ch)

            if not head.nodes[idx]:
                head.nodes[idx] = TrieNode(ch)

            head = head.nodes[idx]

        head.is_end = True


    def search(self, chars):
        head = self.head
        counts = [0]

        for i in range(len(chars)):
            ch = chars[i]
            idx = self._ch_to_idx(ch)

            if not head.nodes[idx]:
                break

            if head.nodes[idx].is_end:
                counts.append(i + 1)

            head = head.nodes[idx]

        return counts


class Solution:
    def minExtraChar(self, s, dictionary):
        trie = Trie()
        n = len(s)

        for w in dictionary:
            trie.insert(w)

        dp = [0]
        for i in range(n):
            dp.append(i + 1)

        for i in range(n):
            skips = trie.search(s[i:])

            for skip in skips:
                # Update the dp table
                dp[i + skip] = min(dp[i + skip], dp[i])

            # Update the current dp table
            # Note: the current dp table is dp[i + 1], not dp[i]
            dp[i + 1] = min(dp[i + 1], dp[i] + 1)

        return dp[-1]
```
