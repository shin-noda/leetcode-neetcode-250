```python
class Solution:
    def longestIncreasingPath(self, matrix):
        rows, cols = len(matrix), len(matrix[0])
        dp = [[-1] * cols for _ in range(rows)]
        self.max_val = 0


        def dfs(r, c):
            # Base Case
            if dp[r][c] != -1:
                return dp[r][c]

            max_from_here = 1

            # Check all 4 directions
            for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
                nr, nc = r + dr, c + dc

                # Boundary check + checking if the neighbour is strictly greater
                if (
                    0 <= nr < rows and
                    0 <= nc < cols and
                    matrix[nr][nc] > matrix[r][c]
                ):
                    max_from_here = max(max_from_here, 1 + dfs(nr, nc))

            # Update the dp table with the final absolute answer for this cell
            dp[r][c] = max_from_here
            self.max_val = max(self.max_val, dp[r][c])

            return dp[r][c]


        for r in range(rows):
            for c in range(cols):
                dfs(r, c)

        return self.max_val
```

```python
solution = Solution()
```

```python
matrix = [[9,9,4],[6,6,8],[2,1,1]]

print(solution.longestIncreasingPath(matrix))
```
