```python
import random

class Solution:
    def kClosest(self, points, k):
        # Helper to get squared distance
        def get_dist(point):
            return point[0]**2 + point[1]**2


        def partition(left, right, pivot_idx):
            pivot_dist = get_dist(points[pivot_idx])

            # Move pivot to the end
            points[pivot_idx], points[right] = points[right], points[pivot_idx]

            store_idx = left

            # Move all points with smaller distance to the left
            for i in range(left, right):
                if get_dist(points[i]) < pivot_dist:
                    points[i], points[store_idx] = points[store_idx], points[i]
                    store_idx += 1

            # Move pivot to its final resting place
            points[right], points[store_idx] = points[store_idx], points[right]

            return store_idx


        def select(left, right, k_target):
            if left == right:
                return

            # Random pivot choice helps avoid the O(N^2) worst-case scenario
            pivot_idx = random.randint(left, right)

            # Partition the array aroun d the pivot
            pivot_idx = partition(left, right, pivot_idx)

            if k_target == pivot_idx:
                return

            elif k_target < pivot_idx:
                # Target is in the left partition
                select(left, pivot_idx - 1, k_target)

            else:
                # Target is in the right partition
                select(pivot_idx + 1, right, k_target)

        # We want to find the boundary for the k-th element (index k - 1)
        select(0, len(points) - 1, k - 1)

        # The first k elements are now guaranteed to be the k closest points
        return points[:k]
```
