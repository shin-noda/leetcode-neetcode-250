```python
class Solution:
    def accountsMerge(self, accounts):
        account_map = {}

        for account in accounts:
            name = account[0]
            emails = set(account[1:])

            if name not in account_map:
                account_map[name] = []

            account_map[name].append(emails)

        # We merge accounts if there are any shared emails among accounts
        for name in account_map:
            # We check if other emails can be merged too.
            first = 0
            while first < len(account_map[name]) - 1:
                second = first + 1
                while second < len(account_map[name]):
                    e1 = account_map[name][first]
                    e2 = account_map[name][second]

                    if not e1.isdisjoint(e2):
                        account_map[name][first].update(e2)
                        account_map[name].pop(second)

                        # Reset first and second
                        first = 0
                        second = 1

                    else:
                        second += 1

                # Increment first
                first += 1

        results = []
        for name in account_map:
            for i in range(len(account_map[name])):
                emails = list(account_map[name][i])
                emails.sort()
                emails.insert(0, name)

                results.append(emails)

        return results
```
