```python
class TimeMap:

    def __init__(self):
        self.time_map = {}


    def set(self, key, value, timestamp):
        if key not in self.time_map:
            self.time_map[key] = {}

        self.time_map[key][timestamp] = value


    def get(self, key, timestamp):
        if key not in self.time_map:
            return ""

        else:
            keys = list(self.time_map[key].keys())

            # Sort keys: O(NlogN)
            left = 0
            right = len(keys) - 1

            # Check if the value exists at the time stamp
            if keys[left] > timestamp:
                return ""

            while left <= right:
                mid = (left + right) // 2

                if keys[mid] <= timestamp:
                    left = mid + 1

                else:
                    right = mid - 1

            return self.time_map[key][keys[right]]
```
