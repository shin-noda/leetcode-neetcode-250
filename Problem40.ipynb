```python
class Solution:
    def combinationSum2(self, candidates, target):
        self.answers = set()
        self.tmp = []

        def backtrack(index, sum_val):
            # Base case and pruning
            if index >= len(candidates) or sum_val >= target:
                if sum_val == target:
                    self.answers.add(tuple(self.tmp))

                return

            last_choice = -1
            for i in range(index, len(candidates)):
                if last_choice == candidates[i]:
                    continue

                else:
                    last_choice = candidates[i]

                # Choose
                self.tmp.append(candidates[i])

                # Explore
                backtrack(i + 1, sum_val + candidates[i])

                # Unchoose
                self.tmp.pop()

        # Early exit
        if sum(candidates) < target:
            return []

        # Sort the list first to avoid the same elements with different order problem.
        candidates.sort()

        # Trim the list because if candidates[i] > target, it'll never be the answer
        last = len(candidates)
        pointer = 0
        while pointer < len(candidates):
            if candidates[pointer] > target:
                last = pointer + 1
                break

            pointer += 1

        candidates = candidates[:last]

        backtrack(0, 0)

        # Convert a set to a list
        return [list(x) for x in self.answers]
```
