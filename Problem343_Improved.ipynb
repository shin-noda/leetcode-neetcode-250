```python
class Solution:
    def integerBreak(self, n):
        # Base cases: handled explicitly because they must be broken into k >= 2 pieces
        if n == 2:
            return 1

        if n == 3:
            return 2

        # dp[i] will store the max product for a component of length i
        dp = [0] * (n + 1)

        # Base pieces that don't need to be broken down further
        dp[1] = 1
        dp[2] = 2
        dp[3] = 3

        for i in range(4, n + 1):
            # We only ever need to check if taking a 2 or a 3 is better
            dp[i] = max(2 * dp[i - 2], 3 * dp[i - 3])

        return dp[n]
```

```python
solution = Solution()
```

```python
n = 2

print(solution.integerBreak(n))
```

```python
n = 10

print(solution.integerBreak(n))
```
