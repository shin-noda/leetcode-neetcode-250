```python
class MyCircularQueue:
    # The Core idea of this solution:
    # We are not actually moving elements in memory.
    # The queue moves logically by moving head, not by shifting elements.
    # And it's circular because whenever we go past the end, we
    # wrap around using % k.

    def __init__(self, k):
        # We pre-allocate the space, just like in a systems environment
        self.queue = [0] * k
        self.k = k
        self.head = 0
        self.size = 0


    def enQueue(self, value):
        if self.isFull():
            return False

        # The Rear index is calculated based on current head and size
        # (head + size) % k gives us the next available empty slot
        target_idx = (self.head + self.size) % self.k
        self.queue[target_idx] = value
        self.size += 1

        return True


    def deQueue(self):
        if self.isEmpty():
            return False

        # To remove, we simply shift the head pointer forward
        self.head = (self.head + 1) % self.k
        self.size -= 1

        return True


    def Front(self):
        if self.isEmpty():
            return -1

        return self.queue[self.head]


    def Rear(self):
        if self.isEmpty():
            return -1

        # Rear is always at (head + size - 1) % k
        rear_idx = (self.head + self.size - 1) % self.k
        return self.queue[rear_idx]


    def isEmpty(self):
        return self.size == 0


    def isFull(self):
        return self.size == self.k
```
