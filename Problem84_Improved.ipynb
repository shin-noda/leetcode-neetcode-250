```python
class Solution:
    def largestRectangleArea(self, heights):
        # Add a 0 at the end to force the stack to clear at the finish
        heights.append(0)

        # Stores (index, height)
        stack = []
        max_area = 0

        for i, h in enumerate(heights):
            start = i

            # If current height is lower, we've found the right boundary
            # for the taller bars on the stack
            while stack and stack[-1][1] > h:
                index, height = stack.pop()
                max_area = max(max_area, height * (i - index))

                # The taller bar's territory can be taken over
                # by the current shorter bar
                start = index

            stack.append((start, h))

        return max_area
```

```python
solution = Solution()
```
