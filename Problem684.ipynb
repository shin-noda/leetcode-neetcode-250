```python
class Solution:
    def findRedundantConnection(self, edges):
        # Idea:
        # Use the tortoise and the hare algorithm

        # On a second though, no.

        # Use this idea.
        # Find edges which has the max connections and remove the last one.
        # If all nodes form a donut, return the last edge

        # New Idea:
        # Trim all leaf nodes until it forms a donut.


        adj = {}
        for a, b in edges:
            if a not in adj:
                adj[a] = set()

            if b not in adj:
                adj[b] = set()

            adj[a].add(b)
            adj[b].add(a)

        leaves = set()
        loop = set()
        max_val = 2
        for key in adj:
            if len(adj[key]) == 1:
                leaves.add(key)

            else:
                if len(adj[key]) > max_val:
                    max_val = len(adj[key])

                loop.add(key)

        # If max_val is 2, it's a donut
        if max_val == 2:
            return edges[-1]

        # We eliminate leaf nodes until it forms a donut
        while leaves:
            leaf = leaves.pop()

            next = adj[leaf].pop()

            # Delete the leaf node from the dict
            adj.pop(leaf)

            # Remove leaf from the loop if it's in there
            if leaf in loop:
                loop.remove(leaf)

            # We remove 'leaf' in the key 'next'
            if next in adj and leaf in adj[next]:
                adj[next].remove(leaf)

                if len(adj[next]) <= 1:
                    leaves.add(next)

        result = []
        # We iterate backwards to find the first edge to be removed
        for i in range(len(edges) - 1, -1, -1):
            if edges[i][0] in loop and edges[i][1] in loop:
                result = edges[i]
                break

        return result
```
