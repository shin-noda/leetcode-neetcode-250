```python
class Solution:
    def canPartition(self, nums):
        # This is a variation of 0/1 Knapsak Problem.
        # Use DP to solve it.

        total_sum = sum(nums)

        # If the total sum is odd, we can't split it equally.
        if total_sum % 2 != 0:
            return False

        target = total_sum // 2

        # DP array: dp[i] will be True if sum 'i' is possible
        # We initialize everything to False, except dp[0] = True
        dp = [False] * (target + 1)

        # Note: By choosing nothing, [], sum = 0.
        dp[0] = True

        # Process each number
        for num in nums:
            # Go backwards from target down to 'num'
            # Going backwards ensures we only use each 'num' once per step

            # Idea:
            # For this one number, update all sums it can contribute to but
            # without letting it reuse itself.
            # Backward loop prevents reuse of the current number in the same iteration.
            for j in range(target, num - 1, -1):
                if dp[j - num]:
                    dp[j] = True

            # Shortcut: If we already found a way to reach the target, stop early
            if dp[target]:
                return True

        return dp[target]
```

```python
solution = Solution()
```

```python
nums = [1,5,11,5]

print(solution.canPartition(nums))
```
