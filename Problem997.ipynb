```python
class UnionFind:
    def __init__(self, size):
        self.parent = list(range(size))

class Solution:
    def findJudge(self, n, trust):
        # Idea:
        # We place the judge node in the center of the graph and town people
        # around the judge.
        # All single arrows point to the center (the judge) and from the judge, nothing
        # will come out.

        # Union-Find works?
        # But if 1 is judge and 1 <- 2 <- 3 or somehting similar exists, that doesn't
        # validate the correct structure??

        # THERE WAS A TEST CASE LIKE: [[1,3],[1,4],[2,3],[2,4],[4,3]]
        # USE UNION-FIND

        # On a seconde thought, we don't use Union-Find.
        # Use the process of elimination

        # There has to be at least (n - 1) trust
        if len(trust) < n - 1:
            return -1

        # Edge case
        if n == 1 and len(trust) == 0:
            return 1

        freq = {}
        candidates = set()
        for i in range(n):
            candidates.add(i + 1)

        for src, dst in trust:
            if not (1 <= src <= n) or not (1 <= dst <= n):
                return -1

            # Everyone trusts the town judge except the judge themselves
            if src in candidates:
                candidates.remove(src)

            freq[dst] = freq.get(dst, 0) + 1

        if not candidates or len(candidates) > 1:
            return -1

        # Only one candidate left in the set
        judge = candidates.pop()

        if judge not in freq or freq[judge] != n - 1:
            return -1

        return judge
```
