```python
import heapq

class Solution:
    def getOrder(self, tasks):
        results = []
        ts = []
        min_heap = []
        time = 0

        for i in range(len(tasks)):
            # (enqueueTime, processingTime, index)
            ts.append((tasks[i][0], tasks[i][1], i))

        # Sort ts by the first element (enqueueTime) in non-increasing (descending) order
        ts.sort(key=lambda x: x[0], reverse=True)

        # We set the time to the smallest enqueTime so that there is at least one
        # executable tasks
        time = ts[-1][0]

        while ts or min_heap:
            while ts and ts[-1][0] <= time:
                enqueueTime, processingTime, index = ts.pop()

                # All elements in this min_heap is executable tasks, so
                # we sort by the processing time and next index.
                # enqueueTime is not important here because all tasks can be
                # processed now.
                heapq.heappush(min_heap, (processingTime, index))

            # Pop the task with the least processingTime and if there is a time, with
            # smaller index
            processingTime, index = heapq.heappop(min_heap)
            results.append(index)

            # Advance the clock
            time += processingTime

            # If there is no element in min_heap, we many need to advance the clock
            if not min_heap and ts:
                time = max(time, ts[-1][0])

        return results
```
