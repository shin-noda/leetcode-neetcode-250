```python
from collections import deque

class Solution:
    def canReach(self, s, minJump, maxJump):
        # Use BFS
        # -> Too slow
        # Use Max Heap
        # -> Not

        # New Idea
        # Create a jump platform that we can jump from
        # and from the dp table, if the platform exists, we can jump

        # Quick Check
        if s[-1] == '1':
            return False

        n = len(s)
        dp = [False] * n

        # Mark dp[0] as True because we start from here
        dp[0] = True

        # Create a platform for O(1) lookup
        # The max width of the platform is (maxJump - minJump) and
        # if there are s[i] == '1', there are holes in the platform
        platform = deque()

        # Jump from the platform
        for i in range(minJump, n):
            # Add a new right edge to the platform
            if s[i - minJump] == '0' and dp[i - minJump]:
                platform.append(i - minJump)

            # Remove the left edge of the platform
            if platform and platform[0] < i - maxJump:
                platform.popleft()

            if platform and s[i] == '0':
                dp[i] = True

        return dp[n - 1]
```

```python
solution = Solution()
```

```python
s = "011010"
minJump = 2
maxJump = 3

print(solution.canReach(s, minJump, maxJump))
```
