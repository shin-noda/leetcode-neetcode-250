```python
class Solution:
    def pacificAtlantic(self, heights):
        p_set = set()
        a_set = set()

        # Initialize the left and rigth edges.
        for i in range(len(heights)):
            p_set.add((i, 0))
            a_set.add((i, len(heights[i]) - 1))

        # Initialize the top and the bottom edges
        for i in range(len(heights[0])):
            p_set.add((0, i))
            a_set.add((len(heights) - 1, i))

        # Create two copies of heights
        p_heights = heights[:]
        a_heights = heights[:]


        def findPacific(r, c, from_r, from_c, path):
            # Base case
            if (
                not (0 <= r <= len(p_heights) - 1) or
                not (0 <= c <= len(p_heights[r]) - 1) or
                p_heights[r][c] > p_heights[from_r][from_c] or
                (r, c) in path
            ):
                return

            # It reached the edge
            if (r, c) in p_set:
                # Add the path to p_set
                p_set.update(path)
                return

            # Update the path
            path.add((r, c))

            # Top, Bottom, Left and Right
            findPacific(r - 1, c, r, c, path.copy())
            findPacific(r + 1, c, r, c, path.copy())
            findPacific(r, c - 1, r, c, path.copy())
            findPacific(r, c + 1, r, c, path.copy())


        def findAtlantic(r, c, from_r, from_c, path):
            # Base case
            if (
                not (0 <= r <= len(a_heights) - 1) or
                not (0 <= c <= len(a_heights[r]) - 1) or
                a_heights[r][c] > a_heights[from_r][from_c] or
                (r, c) in path
            ):
                return

            # It reached the edge
            if (r, c) in a_set:
                # Add the path to p_set
                a_set.update(path)
                return

            # Update the path
            path.add((r, c))

            # Top, Bottom, Left and Right
            findAtlantic(r - 1, c, r, c, path.copy())
            findAtlantic(r + 1, c, r, c, path.copy())
            findAtlantic(r, c - 1, r, c, path.copy())
            findAtlantic(r, c + 1, r, c, path.copy())


        for r in range(len(heights)):
            for c in range(len(heights[r])):
                findPacific(r, c, r, c, set())
                findAtlantic(r, c, r, c, set())

        # Intersection of two sets
        ocean_set = p_set & a_set
        return [list(x) for x in ocean_set]
```
