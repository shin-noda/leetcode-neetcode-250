```python
class Solution:
    def reverse(self, x):
        if x == 0:
            return 0

        r = 0
        s_x = list(str(x))
        n = len(s_x)
        if x > 0:
            for i in range(n // 2):
                s_x[i], s_x[n - 1 - i] = s_x[n - 1 - i], s_x[i]

        else:
            for i in range(1, n // 2 + 1):
                s_x[i], s_x[n - i] = s_x[n - i], s_x[i]

        r = int("".join(s_x))

        if r > 2**31 - 1 or r < -2**31:
            return 0

        return r
```

```python
solution = Solution()
```

```python
x = 123

print(solution.reverse(x))
```
