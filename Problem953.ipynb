```python
class Solution:
    def isAlienSorted(self, words, order):
        alien_dict = {}
        for i in range(len(order)):
            alien_dict[order[i]] = i

        words_list = []
        for w in words:
            words_list.append(list(w))

        for i in range(len(words_list)):
            for j in range(len(words_list[i])):
                words_list[i][j] = alien_dict[words_list[i][j]]

        sorted_list = sorted(words_list)

        return words_list == sorted_list
```
