```python
class Solution:
    def carFleet(self, target, position, speed):
        # O(NlogN)
        # Pair them, sort by position descending
        cars = sorted(zip(position, speed), reverse=True)
        stack = []

        for p, s in cars:
            time = (target - p) / s

            # If stack is empty or this car takes longer than the car in front
            if not stack or time > stack[-1]:
                stack.append(time)

        # The number of unique arrival times in the stack is the number of fleets
        return len(stack)
```

```python
solution = Solution()
```

```python
target = 12
position = [10,8,0,5,3]
speed = [2,4,1,1,3]

print(solution.carFleet(target, position, speed))
```
