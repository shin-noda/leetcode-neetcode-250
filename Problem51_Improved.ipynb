```python
class Solution:
    def solveNQueens(self, n):
        cols = set()

        # r + c
        pos_diag = set()

        # r - c
        neg_diag = set()

        results = []
        board = [["."] * n for _ in range(n)]


        def backtrack(r):
            # Base Case
            if r == n:
                # Compress and collect the board state
                results.append(["".join(row) for row in board])
                return

            for c in range(n):
                # Pruning
                if c in cols or (r + c) in pos_diag or (r - c) in neg_diag:
                    continue

                # Choose
                board[r][c] = 'Q'
                cols.add(c)
                pos_diag.add(r + c)
                neg_diag.add(r - c)

                # Explore
                backtrack(r + 1)

                # Unchoose
                board[r][c] = '.'
                cols.remove(c)
                pos_diag.remove(r + c)
                neg_diag.remove(r - c)

        backtrack(0)

        return results
```
