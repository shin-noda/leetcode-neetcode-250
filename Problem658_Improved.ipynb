```python
class Solution:
    def findClosestElements(self, arr, k, x):
        def binarySearch(target):
            left, right = 0, len(arr) - 1

            while left <= right:
                mid = (left + right) // 2

                if arr[mid] == target:
                    return mid

                elif arr[mid] < target:
                    left = mid + 1

                else:
                    right = mid - 1

            return left

        # Find the insertion point
        index = binarySearch(x)

        # Initialize pointers
        # left is the colset element <= x, right is the next one over
        left = index - 1
        right = index

        # Expand outward k times
        while k > 0:
            # If right is out of bounds, we must take from the left
            if right >= len(arr):
                left -= 1

            # If left is out of bounds, we must take from the right
            elif left < 0:
                right += 1

            # Comparison logic with the tie-breaker
            else:
                if abs(arr[left] - x) <= abs(arr[right] - x):
                    left -= 1

                else:
                    right += 1

            k -= 1

        # Return the range (left is exclusive, right is exclusive)
        return arr[left + 1: right]
```

```python
solution = Solution()
```

```python
arr = [1,2,3,4,5]
k = 4
x = 3

print(solution.findClosestElements(arr, k, x))
```
