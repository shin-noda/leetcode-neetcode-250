```python
class Solution:
    def sortColors(self, nums):
        """
        Do not return anything, modify nums in-place instead.
        """

        n = len(nums)

        last_zero = 0
        last_two = n - 1

        i = 0
        while i < n:
            if nums[i] == 0:
                # We put it at the front
                nums[i], nums[last_zero] = nums[last_zero], nums[i]
                last_zero += 1

            i += 1

        i = n - 1
        while i >= last_zero:
            if nums[i] == 2:
                # We put it at the end
                nums[i], nums[last_two] = nums[last_two], nums[i]
                last_two -= 1

            i -= 1
```

```python
solution = Solution()
```

```python
nums = [2,0,2,1,1,0]

print(solution.sortColors(nums))
```
