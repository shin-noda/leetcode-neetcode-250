```python
from collections import deque

class MyStack:

    def __init__(self):
        # Note: this only behaves as a queue
        self.queue = deque()


    def push(self, x):
        self.queue.append(x)


    def pop(self):
        n = len(self.queue)

        # We "rotate" the elements of queue
        # and We want to leave the last element of queue
        for i in range(n - 1):
            self.queue.append(self.queue.popleft())

        last_val = self.queue.popleft()

        # We bring back the original order
        for i in range(n - 1):
            self.queue.append(self.queue.popleft())

        return last_val


    def top(self):
        n = len(self.queue)

        # We want to leave the last element of queue
        for i in range(n - 1):
            self.queue.append(self.queue.popleft())

        # We record the last element
        last_val = self.queue.popleft()

        # We push it to tmp
        self.queue.append(last_val)

        return last_val


    def empty(self):
        return len(self.queue) == 0
```
