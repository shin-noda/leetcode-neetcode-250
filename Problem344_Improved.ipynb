```python
class Solution:
    def reverseString(self, s):
        """
        Do not return anything, modify s in-place instead.
        """

        # Note: we don't need to return the string

        left = 0
        right = len(s) - 1

        # We don't have to thing about the case when left == right because
        # it's the middle when the length of s is odd, so no need to swap.
        while left < right:
            s[left], s[right] = s[right], s[left]

            left += 1
            right -= 1
```

```python
solution = Solution()
```

```python
s = ["h","e","l","l","o"]

print(solution.reverseString(s))
```

```python
s = ["H","a","n","n","a","h"]

print(solution.reverseString(s))
```
