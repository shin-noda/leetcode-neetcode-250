```python
class Solution:
    def findTargetSumWays(self, nums, target):
        # Idea:
        # This problem is a variation of Last Stone Weight II.
        # This is a 0/1 Knapsack problem.
        # Think about putting all items in two knapsacks.
        # Knapsack A for positive and Knapsack B for negative numbers.
        #
        # Outer loop should be nums and processed only once.
        # The order doesn't matter, so we don't have to sort them.
        #
        # Actually, since nums.lengh <= 20 by the constraint, we can put
        # +num0, -num0, +num1, -num1, ... +numN, -numN
        # and initialized dp with dp[target] = 0.
        # (2^20 = 1048576 -> about 10^6)
        #
        # This problem is like a T/F table.
        # For example, if there are three elements in nums:
        # First column: ++++----
        # Second column: ++--++--
        # Third column: +-+-+-+-
        #
        # The dp stores the frequency of the sum so far.

        # Use a hashmap/dict for dp.
        # dp[i] represents how many combinations sum up to 'i' if dp[i] exists
        dp = {}

        # Initialize dp[num0], dp[-num0] with 1.
        if nums[0] == 0:
            # This is for +0 and -0.
            dp[0] = 2

        else:
            dp[nums[0]] = 1
            dp[-nums[0]] = 1

        sum_set = set()
        sum_set.add(nums[0])
        sum_set.add(-nums[0])

        for i in range(1, len(nums)):
            num = nums[i]
            # We don't directly add a new sum to sum_set to avoid
            # calculating a new sum with the value just calculated for this iteration.
            new_sum_set = set()
            n = len(sum_set)

            # We need to use a new dp because we only want to keep the record of
            # last sums for each iteration, not the entire record of frequencies
            # of sum so far.
            new_dp = {}

            for _ in range(n):
                sum_so_far = sum_set.pop()

                positive = sum_so_far + num
                negative = sum_so_far - num

                new_dp[positive] = new_dp.get(positive, 0) + dp[sum_so_far]
                new_dp[negative] = new_dp.get(negative, 0) + dp[sum_so_far]

                new_sum_set.add(positive)
                new_sum_set.add(negative)

            # Update sum_set
            sum_set = new_sum_set

            # Update dp
            dp = new_dp

        # If the final sum does not end up with 'target', it should return 0.
        return dp.get(target, 0)
```

```python
solution = Solution()
```

```python
nums = [1,1,1,1,1]
target = 3

print(solution.findTargetSumWays(nums, target))
```

```python
nums = [1]
target = 1

print(solution.findTargetSumWays(nums, target))
```

```python
nums = [1,1]
target = 0

# It should return 2
print(solution.findTargetSumWays(nums, target))
```

```python
nums = [0,0,0]
target = 0

# It should return 8 (2^3 = 8)
print(solution.findTargetSumWays(nums, target))
```
