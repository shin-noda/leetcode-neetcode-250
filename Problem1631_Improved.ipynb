```python
import heapq

class Solution:
    def minimumEffortPath(self, heights):
        rows, cols = len(heights), len(heights[0])


        # Standard BFS to see if a path exists
        # using only transitions with an absolute difference <= max_allowable_effort
        def can_reach_destination(max_allowable_effort):
            queue = deque([(0, 0)])
            visited = [[False] * cols for _ in range(rows)]
            visited[0][0] = True

            directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

            while queue:
                r, c = queue.popleft()

                if r == rows - 1 and c == cols - 1:
                    return True

                for dr, dc in directions:
                    nr, nc = r + dr, c + dc

                    if 0 <= nr < rows and 0 <= nc < cols and not visited[nr][nc]:
                        if abs(heights[r][c] - heights[nr][nc]) <= max_allowable_effort:
                            visited[nr][nc] = True
                            queue.append((nr, nc))

            return False

        # Binary Search Space
        left, right = 0, 10**6
        answer = right

        while left <= right:
            mid = (left + right) // 2

            if can_reach_destination(mid):
                # Record this valid effort
                answer = mid

                # Try to find a smaller valid effort
                right = mid - 1

            else:
                # This effort was too small, so increase it
                left = mid + 1

        return answer
```
