```python
class Solution:
    def numSquares(self, n):
        # dp[i] will store the min perfect squares to get to i
        dp = [float('inf')] * (n + 1)

        # Base case: 0 takes 0 squares
        dp[0] = 0

        for i in range(1, n + 1):
            j = 1

            # Only check subtractions of perfect squares
            while j * j <= i:
                dp[i] = min(dp[i], dp[i - j * j] + 1)
                j += 1

        return dp[n]
```

```python
solution = Solution()
```

```python
n = 12

print(solution.numSquares(n))
```
