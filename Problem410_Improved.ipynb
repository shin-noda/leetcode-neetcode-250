```python
class Solution:
    def splitArray(self, nums, k):
        # The smallest possible largest sum must be at least the biggest number
        # The largest possible largest sum is the sum of everything
        left = max(nums)
        right = sum(nums)
        min_val = right

        while left <= right:
            mid = (left + right) // 2

            # Helpter: Can we split nums into k groups with sum <= mid
            groups = 1
            curr_sum = 0
            for num in nums:
                if curr_sum + num > mid:
                    groups += 1
                    curr_sum = num

                else:
                    curr_sum += num

            # If we used k or fewer groups, mid is a potential answer
            if groups <= k:
                min_val = mid

                # Try to find a smaller maximum
                right = mid - 1

            else:
                # Mid is too small, so we ned more groups than allowed
                left = mid + 1

        return min_val
```

```python
solution = Solution()
```
