```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None
```

```python
class Solution:
    def lowestCommonAncestor(self, root, p, q):
        self.lca = None

        def dfs(node):
            if not node or self.lca:
                return 0

            counter = 0
            if node.val == p.val or node.val == q.val:
                counter += 1

            left = dfs(node.left)
            right = dfs(node.right)

            if (counter + left + right) >= 2:
                self.lca = node

            return max(counter, left, right)

        dfs(root)

        return self.lca
```
