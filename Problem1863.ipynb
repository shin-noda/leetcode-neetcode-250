```python
class Solution:
    def subsetXORSum(self, nums):
        self.result = 0

        def backtrack(index, val):
            for i in range(index, len(nums)):
                # Choose
                new_val = val ^ nums[i]
                self.result += new_val

                # Explore
                backtrack(i + 1, new_val)

                # Unchoose
                # No further action required here.

        backtrack(0, 0)

        return self.result
```
