```python
class Solution:
    def search(self, nums, target):
        left, right = 0, len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                return True

            # The flat case: we can't tell which side is sorted.
            # So, just shrink the search space by 1 and try again.
            if nums[left] == nums[mid] == nums[right]:
                left += 1
                right -= 1

            # Left side is sorted
            elif nums[left] <= nums[mid]:
                if nums[left] <= target < nums[mid]:
                    right = mid - 1

                else:
                    left = mid + 1

            # Right side is sorted
            else:
                if nums[mid] < target <= nums[right]:
                    left = mid + 1

                else:
                    right = mid - 1

        return False
```

```python
solution = Solution()
```

```python
nums = [2,5,6,0,0,1,2]
target = 0

print(solution.search(nums, target))
```
