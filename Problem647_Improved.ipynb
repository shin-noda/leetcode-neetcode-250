```python
class Solution:
    def countSubstrings(self, s):
        counts = 0

        def expand_around_centre(left, right):
            result = 0
            while left >= 0 and right < len(s) and s[left] == s[right]:
                result += 1
                left -= 1
                right += 1

            return result


        for i in range(len(s)):
            # Odd lengh palindromes (e.g., "aba")
            counts += expand_around_centre(i, i)

            # Even length palindromes (e.g., "abba")
            counts += expand_around_centre(i, i + 1)

        return counts
```
