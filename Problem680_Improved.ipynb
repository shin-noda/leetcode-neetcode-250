```python
class Solution:
    def validPalindrome(self, s):
        # Use Two Pointer methods and a helper method
        def is_palindrome(left, right):
            while left < right:
                if s[left] != s[right]:
                    return False

                left += 1
                right -= 1

            return True

        left, right = 0, len(s) - 1

        while left < right:
            if s[left] == s[right]:
                left += 1
                right -= 1

            else:
                # Either skip s[left] or skip[right]
                # If either one results in a palindrome, it's true
                return is_palindrome(left + 1, right) or is_palindrome(left, right - 1)

        return True
```

```python
solution = Solution()
```
