```python
class Solution:
    def findClosestElements(self, arr, k, x):
        def binarySearch(target):
            left = 0
            right = len(arr) - 1

            while left <= right:
                mid = (left + right) // 2

                if arr[mid] == x:
                    return mid

                elif arr[mid] < x:
                    left = mid + 1

                else:
                    right = mid - 1

            return left

        # Check if arr has only one element
        if len(arr) <= 1:
            return arr

        index = binarySearch(x)
        results = []

        if index < 0:
            index = 0

        elif index >= len(arr):
            index = len(arr) - 1

        # Use Two Pointer method
        left = index
        right = index

        if index == 0:
            right = index + 1

        else:
            left -= 1

        while len(results) < k:
            diff1 = abs(arr[left] - x)
            diff2 = abs(arr[right] - x)

            if diff1 <= diff2:
                # Note: if diff1 == diff2, we choose left as a tie-breaker
                results.append(arr[left])
                left -= 1

            # This is for the case diff1 > diff2
            else:
                results.append(arr[right])
                right += 1

            # There is no elements left of the leftside
            if left < 0:
                while len(results) < k:
                    results.append(arr[right])
                    right += 1

            # There is no elements left on the right side
            if right >= len(arr):
                while len(results) < k:
                    results.append(arr[left])
                    left -= 1

        # O(NlogN)
        results.sort()

        return results
```

```python
solution = Solution()
```
