```python
class Solution:
    def checkIfPrerequisite(self, numCourses, prerequisites, queries):
        # DFS / BFS with Memoization

        # Build the graph
        adj = {i: [] for i in range(numCourses)}
        for prereq, course in prerequisites:
            adj[prereq].append(course)

        # Map each course to all its reachable descendants
        # reach[i] will store a set of all courses that require course i
        reach = {}

        def dfs(node):
            # Base Case
            if node in reach:
                return reach[node]

            # Start with immediate neighbours
            results = set(adj[node])
            for neighbour in adj[node]:
                # Union with everything the neighbour can reach
                results.update(dfs(neighbour))

            reach[node] = results
            return results


        # Run DFS for every node to fully populate the reach map
        for i in range(numCourses):
            dfs(i)

        # Answer queries in O(1) time
        return [dst in reach[src] for src, dst in queries]
```
