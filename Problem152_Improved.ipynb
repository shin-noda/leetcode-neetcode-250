```python
class Solution:
    def maxProduct(self, nums):
        # Idea:
        # If I must end my subarray here, what is the best and worst product I can have?

        result = max(nums)
        curr_max, curr_min = 1, 1

        for num in nums:
            if num == 0:
                # Rest the products, but keep result updated
                curr_max, curr_min = 1, 1
                continue

            # If num is negative, multiplying swaps the max and min values
            tmp = curr_max
            curr_max = max(num * curr_max, num * curr_min, num)
            curr_min = min(num * tmp, num * curr_min, num)

            result = max(result, curr_max)

        return result
```

```python
solution = Solution()
```
