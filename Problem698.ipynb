```python
class Solution:
    def canPartitionKSubsets(self, nums, k):
        # Quick check
        nums.sort(reverse=True)
        sum_val = sum(nums)
        max_val = nums[0]
        target = sum_val // k

        if sum_val % k != 0 or max_val > target:
            return False

        # We want all elements in 'buckts' to be the target value.
        buckets = [0] * k
        self.is_achievable = False

        def backtrack(count):
            # Base case
            if count >= len(nums):
                self.is_achievable = True
                return

            for i in range(k):
                # Choose
                buckets[i] += nums[count]

                # Explore
                if not self.is_achievable and buckets[i] <= target:
                    backtrack(count + 1)

                # Backtrack
                buckets[i] -= nums[count]

                # Pruning: if buckets[i] is 0, which means it doens't work for other i.
                if buckets[i] == 0:
                    break

        backtrack(0)

        return self.is_achievable
```
