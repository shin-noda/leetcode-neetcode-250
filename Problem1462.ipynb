```python
class Solution:
    def checkIfPrerequisite(self, numCourses, prerequisites, queries):
        results = []
        node_map = {}

        # Map all nodes
        for src, dst in prerequisites:
            if src not in node_map:
                node_map[src] = set()

            node_map[src].add(dst)

        # Check if path exists
        for src, dst in queries:
            # Check if src exists in the map.
            # If it doesn't exist, srt and dst are not connected.
            if src not in node_map:
                results.append(False)
                continue

            is_accessible = False

            # Use a set for O(1) lookup
            visited = set()
            stack = {src}

            while stack and not is_accessible:
                n = len(stack)
                tmp = set()

                for _ in range(n):
                    node = stack.pop()

                    if node == dst:
                        is_accessible = True
                        break

                    if node not in visited and node in node_map:
                        visited.add(node)

                        tmp.update(node_map[node])

                # Update the stack
                stack = tmp

            results.append(is_accessible)

        return results
```
