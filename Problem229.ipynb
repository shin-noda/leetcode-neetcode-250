```python
class Solution:
    def majorityElement(self, nums):
        n = len(nums)
        n_3 = n // 3
        freq = {}
        results = []

        for num in nums:
            freq[num] = freq.get(num, 0) + 1

        for key in freq:
            value = freq[key]

            if value > n_3:
                results.append(key)

        return results
```

```python
solution = Solution()
```

```python
nums = [3,2,3]

print(solution.majorityElement(nums))
```

```python
nums = [1]

print(solution.majorityElement(nums))
```
