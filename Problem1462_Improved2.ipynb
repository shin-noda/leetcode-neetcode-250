```python
class Solution:
    def checkIfPrerequisite(self, numCourses, prerequisites, queries):
        # Use the Floyd-Warshall Algorithm
        # We can use it here because V <= 100 and O(V^3) is 100^3 = 10^6

        # grid[i][j] means i is a prerequisite of j
        grid = [[False] * numCourses for _ in range(numCourses)]

        # Set up direct prerequisites
        for prereq, course in prerequisites:
            grid[prereq][course] = True

        # Floyd-Warshall: For every intermediate node k,
        # check if i can reach j through k
        for k in range(numCourses):
            for i in range(numCourses):
                for j in range(numCourses):
                    if grid[i][k] and grid[k][j]:
                        grid[i][j] = True

        # Now each query is a O(1) lookup
        return [grid[src][dst] for src, dst in queries]
```
