```python
class Solution:
    def combinationSum2(self, candidates, target):
        # Sort to group duplicates and enable ordered pruning
        candidates.sort()
        results = []

        def backtrack(index, remaining, current_combo):
            # Base Case: we found a valid combination
            if remaining == 0:
                results.append(list(current_combo))
                return

            for i in range(index, len(candidates)):
                # Core Pruning: if the current number exceeds the remaining target,
                # no number after it will work either because the list is sorted
                if candidates[i] > remaining:
                    break

                # Duplicate Skipping: Skip identical numbers at the same decision level
                if i > index and candidates[i] == candidates[i - 1]:
                    continue

                # Make the choice
                current_combo.append(candidates[i])

                # Move to the next index with the updated remaining target
                backtrack(i + 1, remaining - candidates[i], current_combo)

                # Undo the choice
                current_combo.pop()


        backtrack(0, target, [])

        return results
```
