```python
class Solution:
    def validPalindrome(self, s):
        # Use Two Pointer methods
        left = 0
        right = len(s) - 1

        while left < right:
            # Match
            if s[left] == s[right]:
                left += 1
                right -= 1

            # Mismatch
            else:
                is_matched = True

                # We create backups
                tmp_left = left
                tmp_right = right

                # We check one right element of the Left Pointer
                if s[tmp_left + 1] == s[tmp_right]:
                    # We move Left twice to the left
                    left = tmp_left + 2
                    right = tmp_right - 1

                    while left < right:
                        # Match
                        if s[left] == s[right]:
                            left += 1
                            right -= 1

                        # Mismatch
                        else:
                            is_matched = False
                            break

                    if is_matched:
                        return True

                # We check out left element of the Right Pointer
                if s[tmp_left] == s[tmp_right - 1]:
                    # We move Right twice to the right
                    right = tmp_right - 2
                    left = tmp_left + 1

                    while left < right:
                        # Match
                        if s[left] == s[right]:
                            left += 1
                            right -= 1

                        # Mismatch
                        else:
                            return False

                # We cannot make it palindrome after deleting one element from s.
                else:
                    return False

        return True
```

```python
solution = Solution()
```
