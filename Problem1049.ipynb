```python
class Solution:
    def lastStoneWeightII(self, stones):
        # Idea:
        # Transforming the problem into 0/1 Knapsack.
        # Let's say the total weight of all stones is total_weight.
        # If we can form a group of stones that gets as close as possible to
        # total_weight / 2, the other group will naturally take the rest, and
        # the difference between them will be minimized.
        # -Knapsack Capacity: total_weight // 2
        # -Items: The stones. Each stone can either be included in our group or left out
        # -Objective: Maximize the weight we can fit into the knapsack without
        # exceeding the capacity.
        #
        # Once we find the maximum weight we can fit into this half-sized knapsack
        # (let's call it target_sum), the remaining stones will weight total_weight - target_sum.
        #
        # The final answer (the smallest leftover stone) will just be the difference
        # between the two groups:
        # Result = (total_weight - target_sum) - target_sum

        # The key insight
        # No matter what sequences of smashes you do, the final stone weight
        # can alwasy be written as:
        # +-stone1+-stone2+-stone3...
        # 8, 7, 4, 2, 1 -> 8 - 7 + 4 - 2 + 1
        # => the order of smashing doesn't really matter anymore.
        # What matter is:
        # Which stones end up with a "+" sign and which stones end up with a "-" sign?

        total_weight = sum(stones)
        target = total_weight // 2

        # DP set to store all possible sums we can generate
        # that are less than or equal to 'target'
        dp = {0}

        for stone in stones:
            # We create a new set of sums by trying to add the current stone
            # to every existing sum we've found so far
            new_sums = set()
            for current_sum in dp:
                if current_sum + stone <= target:
                    new_sums.add(current_sum + stone)

            # Merget the new sums back into our DP tracker
            dp.update(new_sums)

        # The closest sum we could get to total_weight // 2
        max_target_sum = max(dp)

        # The answer is the difference between the two groups
        return (total_weight - max_target_sum) - max_target_sum
```

```python
solution = Solution()
```

```python
stones = [2,7,4,1,8,1]

print(solution.lastStoneWeightII(stones))
```
