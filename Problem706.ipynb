```python
class MyHashMap:

    def __init__(self):
        self.size = 10000
        self.my_list = [[] for _ in range(self.size)]


    def _hash(self, key):
        return key % self.size


    def put(self, key, value):
        hash_idx = self._hash(key)

        if not self.my_list[hash_idx]:
            self.my_list[hash_idx].append([key, value])

        else:
            # This is O(1) because the max size is fixed regardless of inputs
            for i in range(len(self.my_list[hash_idx])):
                k, v = self.my_list[hash_idx][i][0], self.my_list[hash_idx][i][1]

                if k == key:
                    self.my_list[hash_idx][i][1] = value
                    return

            # We add a new [key, value] pair since it does not exist
            self.my_list[hash_idx].append([key, value])


    def get(self, key):
        hash_idx = self._hash(key)

        if not self.my_list[hash_idx]:
            return -1

        else:
            for i in range(len(self.my_list[hash_idx])):
                k, v = self.my_list[hash_idx][i][0], self.my_list[hash_idx][i][1]

                if k == key:
                    return self.my_list[hash_idx][i][1]

        return -1


    def remove(self, key):
        hash_idx = self._hash(key)

        if not self.my_list[hash_idx]:
            return

        else:
            for i in range(len(self.my_list[hash_idx])):
                k, v = self.my_list[hash_idx][i][0], self.my_list[hash_idx][i][1]

                if k == key:
                    # We swap with the last element and pop the last element
                    # so that it's O(1)
                    self.my_list[hash_idx][i], self.my_list[hash_idx][-1] =\
                    self.my_list[hash_idx][-1], self.my_list[hash_idx][i]

                    self.my_list[hash_idx].pop()
                    return
```
