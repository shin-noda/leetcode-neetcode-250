```python
class Solution:
    def combinationSum4(self, nums, target):
        # Idea:
        # 0/1 Knapsack Problem but it process forward
        # because it can use a number multiple times.

        # Another Idea:
        # Let's say nums = [1,2,3], target = 4.
        # To make 1, dp[1] = 1, (1)
        # To make 2, dp[2] = 2, (1,1),(2)
        # To make 3, dp[3] = 4, (1,1,1),(1,2),(2,1),(3)
        # To make 4. dp[4] = 7, (1,1,1,1),(1,1,2),(1,2,1),(1,3),(2,1,1),(2,2),(3,1)

        # dp[2]: (2), (dp[1], dp[1])
        # dp[3]: (3), (dp[1], dp[2]), (dp[2], dp[1])

        # dp[i] stores how many combinations are there to make 'i'
        # For the example above, dp[3] = 4

        # Another Idea:
        # This problem is similar to 91. Decode Ways.
        # For  Decode Ways, at every position, you can arrive there from either:
        # 1. decoding one digit
        # 2. decoding two digits
        #
        # But for this problem, for this example: [1,2,3], target = 4,
        # you can arrive at 4 from
        # 1, 2 and 3.
        # So at i = 4,
        # dp[4] is dp[1] + dp[2] + dp[3]
        # dp[i - nums]

        # Maybe this problem is a variation of a staircase problem.
        # But instead of 1 or 2 steps, it can take n steps where n is an element in nums.

        dp = [0] * (target + 1)

        # This is for O(1) lookup
        num_set = set(nums)

        # Sort nums: O(NlogN)
        nums.sort()

        for i in range(1, target + 1):
            if i in num_set:
                dp[i] += 1

            for num in nums:
                # This is '>=', not '>' because we checked the '=' case
                # for looking at 'num_set' above.
                if num >= i:
                    break

                # You arrive at the i-th step from the (i-num)th step.
                dp[i] += dp[i - num]

        return dp[-1]
```

```python
solution = Solution()
```

```python
nums = [1,2,3]
target = 4

print(solution.combinationSum4(nums, target))
```
