```python
import heapq

class Solution:
    def carPooling(self, trips, capacity):
        # Sort trips strictly by pickup time so we process chronologically
        trips.sort(key=lambda x: x[1])

        # min_heap stores: (dropoff_location, passenger_count)
        min_heap = []

        for num, f, t in trips:
            # Drop off everyone whose destination is <= current pickup location 'f'
            while min_heap and min_heap[0][0] <= f:
                _, ppl = heapq.heappop(min_heap)
                capacity += ppl

            # Board the new passengers
            capacity -= num
            if capacity < 0:
                return False

            # Record their future dropoff
            heapq.heappush(min_heap, (t, num))

        return True
```
