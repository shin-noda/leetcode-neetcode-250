```python
class Solution:
    def isNStraightHand(self, hand, groupSize):
        # Idea:
        # This is a Sliding Windown problem and the size of the window
        # is fixed and groupSize.

        # Find the max value in the list and create a 'countss' table
        # size of (n + 1) so that the value in the array is an index of the list.
        # Also, use the hands as a hashmap but I use a list instead of a hashmap
        # so that I can avoid if key in hashmap each time.

        # I'm not sure how to implement the idea about, so step back and
        # think about a naive approach.

        # Naive approach
        # For the outer loop, we iterate though 'counts' from i = 1 to i = N.
        # In 'counts', when you find an element which is non-zero, start an
        # inner loop starting from there, subtract that value from counts[i]
        # to counts[i + groupSize].
        # If there is not enough element in counts[j] where i < j <= i + groupSize,
        # it returns False.
        # I think this works but the problem is when groupSize is huge like 10^4,
        # the time complexity is O(N^2).

        # How to optimize the naive approach?
        # Clearly, there is a repetitive work because there is an outloop and
        # you scan 'count's multiple times for each time you find non-zero element.

        # Realized that valid examples like this:
        # counts = [0, 1, 2, 2, 1, 0, 1, 1, 1] or
        # counts = [0,1,4,5,4,1]
        # each non-zero segments are palindromic
        # either odd or even sequences.
        # But, sequence like counts = [0,1,4,1] is invalid because
        # hand = [1,2,2,2,2,3] and three 2's left.

        # I think this problem is about compressing information in a faster way.
        # For example, if you have:
        # hand = [1,2,2,2,2,3,3,3,3,3,4,4,4,4,5], groupSize=3, then
        # counts = [0,1,4,5,4,1].
        # Now, how can we "compress" the naive approach so that it's faster?

        # Create a dp table which can refer to the previous element to know
        # how many numbers should be subtracted at dp[i].
        # At dp[i], which dp[x] should it refer to?

        # Maybe dp[i] keeps track of duplicate/overlapping intervals?

        # Maybe think problem like a variation of a staircase prlblem???

        # Forward thinking like at dp[i], check the right edge of the window.
        # We ignore when counts[i] == 0.

        # Maybe sort the hand itself??

        # When the value increases like counts[i] < counts[i + 1],
        # there should be an additional window.

        # At dp[i], it has to carry information how many overlapping windws are there.

        # Actually, the naive approach I'm thinking about is O(N) amortized because
        # each element in 'hand' will be scanned exactly once.

        # I got the Memory Limit Exceeded error :(
        # -> Time to use hashmap


        # O(N)
        max_val = max(hand)

        # Quick Checks: the size of hand should be a multiple of groupSize
        # Also, 'max_val' should be greater than or equal to the groupSize
        if len(hand) % groupSize != 0 or max_val < groupSize - 1:
            return False

        counts = {}

        # O(N)
        for h in hand:
            counts[h] = counts.get(h, 0) + 1

        keys = list(counts.keys())

        # O(NlogN)
        keys.sort()

        # This is a double-loop but every element in 'hand' is
        # only scanned once in this loop, so it's O(N).
        for i in keys:
            loops = counts[i]

            for _ in range(loops):
                for j in range(i, i + groupSize):
                    if j > max_val or j not in counts or counts[j] <= 0:
                        return False

                    counts[j] -= 1

        return True
```
