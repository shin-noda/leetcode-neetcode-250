```python
import heapq

class Solution:
    def networkDelayTime(self, times, n, k):
        # Use Dijkstra's Algorithm
        min_time = float('-inf')

        # Create a dict (node -> connected nodes)
        adj = {}

        # Create a weighted graph
        wg = [float('inf')] * (n + 1)

        # Iterate all times array
        for r, c, w in times:
            if r not in adj:
                adj[r] = set()

            adj[r].add((c, w))

        # (time, current_node)
        min_heap = [(0, k)]
        visited = set()

        # We mark the time of the starting node as 0
        wg[k] = 0

        while min_heap:
            t, curr = heapq.heappop(min_heap)

            # Add the current node as visited
            visited.add(curr)

            # Note: this is the max of those 2, not min.
            # Some nodes can be reached earlier, so we want to know the last tiem
            # that the last node was visited.
            min_time = max(min_time, t)

            # If len(visited) is n, we visited all n nodes
            if len(visited) >= n:
                return min_time

            # If we found a worse path than already recorded, skip it
            if wg[curr] < t:
                continue

            size = 0
            nodes = set()
            if curr in adj:
                nodes = adj[curr]
                size = len(nodes)

            for _ in range(size):
                # new_node, new_time
                nn, nt = nodes.pop()

                # If this path offer a faster time, update and push it
                if wg[nn] > t + nt:
                    wg[nn] = t + nt
                    heapq.heappush(min_heap, (t + nt, nn))

        # Impossible to receive the signal
        return -1
```
