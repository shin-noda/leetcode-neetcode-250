```python
class Solution:
    def rob(self, nums):
        # Check if the number of housrs are less than or equal to 3
        if len(nums) <= 3:
            return max(nums)

        # Idea:
        # There are 3 patters
        # 1. Choose nums[0] and nums[-1] <- illegal
        # 2. Choose nums[0] and not to choose nums[-1] <- legal
        # 3. Choose nums[1] and may choose nums[-1] <- legal
        #
        # Since pattern 1 is illegal, we only need to consider two cases.
        # - starts with index[0] and pop the last element
        # - ignore the initial element and start with index[1]

        nums0 = list(nums[:len(nums) - 1])
        nums1 = list(nums[1:])

        n = len(nums1)

        # We ignore the last element for this version
        dp0 = [0] * n
        dp0[0] = nums0[0]
        dp0[1] = max(nums0[1], nums0[0])

        # We ignore the first element for this version
        dp1 = [0] * n
        dp1[0] = nums1[0]
        dp1[1] = max(nums1[1], nums1[0])

        prev0 = dp0[0]
        prev1 = dp1[0]

        curr0 = dp0[1]
        curr1 = dp1[1]

        for i in range(2, n):
            dp0[i] = max(curr0, nums0[i] + prev0)
            dp1[i] = max(curr1, nums1[i] + prev1)

            prev0 = curr0
            prev1 = curr1

            curr0 = dp0[i]
            curr1 = dp1[i]

        return max(dp0[-1], dp1[-1])
```

```python
solution = Solution()
```
