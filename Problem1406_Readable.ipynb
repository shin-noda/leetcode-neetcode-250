```python
class Solution:
    def stoneGameIII(self, stoneValue):
        n = len(stoneValue)

        # dp[i]: best score difference for the player whose turn it is at index i
        # Extra space avoids bounds checks
        dp = [0] * (n + 1)

        # Fill from right to left
        for i in range(n - 1, -1, -1):
            best = float('-inf')
            current_sum = 0

            # Take 1, 2, or 3 stones
            for k in range(1, 4):
                if i + k <= n:
                    current_sum += stoneValue[i + k - 1]

                    # Note: dp[i + k] is the best socre difference for your opponent
                    best = max(best, current_sum - dp[i + k])

            dp[i] = best

        if dp[0] > 0:
            return "Alice"

        elif dp[0] < 0:
            return "Bob"

        return "Tie"
```

```python
solution = Solution()
```

```python
stoneValue = [1,2,3,7]

print(solution.stoneGameIII(stoneValue))
```
