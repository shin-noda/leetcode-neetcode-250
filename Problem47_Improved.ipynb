```python
class Solution:
    def permuteUnique(self, nums):
        results = []

        def backtrack(i):
            # Base Case: if we've fixed every position, we have a complete permutation
            if i == len(nums):
                # Take a shallow copy of the current state
                results.append(nums[:])
                return

            # Track values we've already placed at index 'i' for this level of recursion
            lookup = set()

            for j in range(i, len(nums)):
                # If we haven't seen this number placed at index 'i' yet
                if nums[j] not in lookup:
                    lookup.add(nums[j])

                    # Swap to put nums[j] at the current position 'i'
                    nums[i], nums[j] = nums[j], nums[i]

                    # Move to the next position
                    backtrack(i + 1)

                    # Undo the swap
                    nums[i], nums[j] = nums[j], nums[i]

        backtrack(0)

        return results
```
