```python
class Solution:
    def characterReplacement(self, s, k):
        count = {}
        max_streak = 0

        left = 0

        # Track the most frequent char in the current window
        max_freq = 0

        for right in range(len(s)):
            # Add the current character to our frequency map
            count[s[right]] = count.get(s[right], 0) + 1

            # Update the most common character in my current window
            max_freq = max(max_freq, count[s[right]])

            # Window length is (right - left + 1)
            # If (length - max_freq) > k, we have too many "other" characters
            while (right - left + 1) - max_freq > k:
                # Note: we don't strictly need to update max_freq here
                # because max_streak only increases when max_freq increases
                count[s[left]] -= 1
                left += 1

            max_streak = max(max_streak, right - left + 1)

        return max_streak
```

```python
solution = Solution()
```
