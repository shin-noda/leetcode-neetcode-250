```python
class Solution:
    def lemonadeChange(self, bills):
        # Each represents the number of $5 and $10 bills.
        # Initially, we don't have any change
        # Note: we don't keep track of $20 bills because we don't need it.
        b_5 = 0
        b_10 = 0

        for bill in bills:
            # No change needed
            if bill == 5:
                b_5 += 1

            elif bill == 10:
                if b_5 <= 0:
                    return False

                b_5 -= 1
                b_10 += 1

            else:
                # We always give customers a $10 bill first because
                # $5 bills have more flexibility
                if b_10 > 0 and b_5 > 0:
                    b_10 -= 1
                    b_5 -= 1

                elif b_5 >= 3:
                    b_5 -= 3

                else:
                    return False

        return True
```

```python
solution = Solution()
```

```python
bills = [5,5,5,10,20]

print(solution.lemonadeChange(bills))
```
