```python
class Solution:
    def findRedundantConnection(self, edges):
        # Every node is initially its own parent (1-indexed)
        parent = [i for i in range(len(edges) + 1)]

        # Find function with path compression
        def find(n):
            # Get the immediate parent of node n
            p = parent[n]

            # Check if p is NOT the parent
            if p != n:
                # Recursively find the parent
                parent[n] = find(p)

            return parent[n]

        # Union function
        def union(n1, n2):
            p1, p2 = find(n1), find(n2)
            if p1 == p2:
                # They are already connected, so we detected the cycle
                return False

            # Link one group to the other
            parent[p1] = p2

            return True

        for u, v in edges:
            if not union(u, v):
                return [u, v]
```
