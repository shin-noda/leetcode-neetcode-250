```python
class Solution:
    def calPoints(self, operations):
        stack = []

        for op in operations:
            # Note: .isdigit() does not work for negative numbers
            if op.isdigit() or op[0] == '-':
                stack.append(int(op))

            elif op == '+':
                stack.append(stack[-1] + stack[-2])

            elif op == 'D':
                stack.append(stack[-1] * 2)

            elif op == 'C':
                stack.pop()

        # Note: if stack is [], sum([]) returns 0.
        return sum(stack)
```

```python
solution = Solution()
```

```python
ops = ["5","2","C","D","+"]

print(solution.calPoints(ops))
```

```python
ops = ["5","-2","4","C","D","9","+","+"]

print(solution.calPoints(ops))
```

```python
ops = ["1","C"]

print(solution.calPoints(ops))
```
