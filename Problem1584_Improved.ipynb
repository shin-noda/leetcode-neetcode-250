```python
import heapq

class Solution:
    def minCostConnectPoints(self, points):
        n = len(points)
        if n <= 1:
            return 0

        total_cost = 0
        visited = set()

        # (weight, current_node)
        min_heap = [(0, 0)]

        # Optimization: stop as soon as all nodes are visited
        while len(visited) < n:
            w, curr = heapq.heappop(min_heap)

            if curr in visited:
                continue

            visited.add(curr)
            total_cost += w

            # Extract coordinates of the current node
            x1, y1 = points[curr]

            # Calculate the Manhattan Distances on the fly
            # instead of using an adjacency matrix
            for next_node in range(n):
                if next_node not in visited:
                    x2, y2 = points[next_node]
                    md = abs(x1 - x2) + abs(y1 - y2)
                    heapq.heappush(min_heap, (md, next_node))

        return total_cost
```
