```python
class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [1] * n
        self.components = n


    def find(self, i):
        if self.parent[i] == i:
            return i

        # Path compression
        self.parent[i] = self.find(self.parent[i])
        return self.parent[i]


    def union(self, i, j):
        root_i = self.find(i)
        root_j = self.find(j)

        if root_i != root_j:
            # Union by rank
            if self.rank[root_i] > self.rank[root_j]:
                self.parent[root_j] = root_i

            elif self.rank[root_i] < self.rank[root_j]:
                self.parent[root_i] = root_j

            else:
                self.parent[root_j] = root_i
                self.rank[root_i] += 1

            self.components -= 1

            return True

        return False


class Solution:
    def canTraverseAllPairs(self, nums):
        n = len(nums)
        if n == 1:
            return True

        uf = UnionFind(n)

        # Maps prime factor -> first array index seen
        prime_to_index = {}

        for i, num in enumerate(nums):
            # If 1 is in the array and there are multiple elements,
            # it can never connect to anything
            if num == 1:
                return False

            # Standard prime factorization in O(sqrt(num))
            d = 2
            val = num
            while d * d <= val:
                if val % d == 0:
                    # d is a prime factor
                    if d in prime_to_index:
                        uf.union(i, prime_to_index[d])

                    else:
                        prime_to_index[d] = i

                    # Remove all instances of this factor
                    while val % d == 0:
                        val //= d

                d += 1

            # If val > 1 reamaining, the leftover value itself is a prime factor
            if val > 1:
                if val in prime_to_index:
                    uf.union(i, prime_to_index[val])

                else:
                    prime_to_index[val] = i

        # If all indices are unified, components should drop down to 1
        return uf.components == 1
```
