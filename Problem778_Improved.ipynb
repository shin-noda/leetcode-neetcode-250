```python
import heapq

class Solution:
    def swimInWater(self, grid):
        # Use Modified Dijkstra's Algorithm
        n = len(grid)

        # (max_elevation_seen_so_far, r, c)
        min_heap = [(grid[0][0], 0, 0)]

        visited = [[False] * n for _ in range(n)]
        visited[0][0] = True

        directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

        while min_heap:
            time, r, c = heapq.heappop(min_heap)

            # Since we use a min-heap, the first time we pop the destination,
            # it is guaranteed to be the minimum possible time
            if r == n - 1 and c == n - 1:
                return time

            for dr, dc in directions:
                nr, nc = r + dr, c + dc

                if 0 <= nr < n and 0 <= nc < n and not visited[nr][nc]:
                    visited[nr][nc] = True

                    # The time to enter the next cell is the max of the
                    # time accumulated so far or the next cell's elevation
                    next_time = max(time, grid[nr][nc])
                    heapq.heappush(min_heap, (next_time, nr, nc))

        return -1
```
