```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
class Solution:
    def inorderTraversal(self, root):
        # Check if root exists
        if not root:
            return []

        # Implement Iterative solution
        results = []

        queue = [(root, 0)]
        while queue:
            n = len(queue)

            for i in range(n):
                node, index = queue.pop()

                results.insert(index, node.val)

                if node.left:
                    queue.append((node.left, index))

                if node.right:
                    queue.append((node.right, index + 1))

        return results
```
