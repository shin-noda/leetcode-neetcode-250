```python
class Solution:
    def reverseString(self, s):
        """
        Do not return anything, modify s in-place instead.
        """

        # Note: we don't need to return the string

        n = len(s)
        for i in range(n // 2):
            # Swap the left and the right elements
            # We need -1 because when i = 0,
            # n - 0 = n which is out of index.
            s[i], s[n - 1 - i] = s[n - 1 - i], s[i]
```

```python
solution = Solution()
```

```python
s = ["h","e","l","l","o"]

print(solution.reverseString(s))
```

```python
s = ["H","a","n","n","a","h"]

print(solution.reverseString(s))
```
