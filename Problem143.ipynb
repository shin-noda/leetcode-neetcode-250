```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
class Solution:
    def reorderList(self, head):
        """
        Do not return anything, modify head in-place instead.
        """

        # Use the tortoise and the hare algorithm to find the mid point
        dummy = ListNode()
        dummy.next = head

        if not head.next or not head.next.next:
            return head

        slow = head
        fast = head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        # We cut the node into half
        half = slow.next
        slow.next = None

        # We need to reverse the right half of the node
        prev = None
        curr = half
        while curr:
            # Have a backup
            tmp = curr.next

            # Connect the current one to the prev
            curr.next = prev

            # Update the prev
            prev = curr

            # Go to the next node
            curr = tmp

        # Now, prev is the starting point of the reverse nodes
        head = dummy.next
        rev = prev

        while head and rev:
            tmp_head = head.next
            tmp_rev = rev.next

            head.next = rev
            rev.next = tmp_head

            head = tmp_head
            rev = tmp_rev

        return dummy.next
```
