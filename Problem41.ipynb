```python
class Solution:
    def firstMissingPositive(self, nums):
        # Use cyclic sort ideas to achieve
        # O(N) time complexity and O(1) space complexity.

        # If the array length is n, the answer must be in this range:
        # [1, n + 1]
        #
        # e.g. if the array length is 3 and the array is [1,2,3],
        # the answer is 4.
        #
        # e.g. if the array length is 3 and the array is [1,2,4],
        # the answer is 3.
        #
        # Therefore, we only care about numbers in this range:
        # 1 <= num <= n
        n = len(nums)

        # Replace irrelevant numbers with n + 1.
        for i in range(n):
            if nums[i] <= 0 or nums[i] > n:
                nums[i] = n + 1

        # Use indexes as a hash map (flip signs to negative).
        # We use indexes to record which numbers exist.
        for i in range(n):
            # Use abs() because numbers may become negative later.
            num = abs(nums[i])

            # Ignore n + 1 because it is irrelevant.
            if num <= n:
                # Compute the correct index for this number.
                #
                # e.g. if num is 1, the correct index is 1 - 1 = 0
                # e.g. if num is 5, the correct index is 5 - 1 = 4
                idx = num - 1

                # Use a negative value as a sign that
                # the number corresponding to this index exists.
                #
                # e.g. if the current array is [3,4,-5,1],
                # it means that 3 exists because nums[2] is negative.
                if nums[idx] > 0:
                    nums[idx] *= -1

        # The first positive index + 1 is the answer.
        for i in range(n):
            if nums[i] > 0:
                return i + 1

        return n + 1
```

```python
solution = Solution()
```

```python
nums = [1,2,0]

print(solution.firstMissingPositive(nums))
```

```python
nums = [3,4,-1,1]

print(solution.firstMissingPositive(nums))
```
