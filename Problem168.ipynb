```python
class Solution:
    def convertToTitle(self, columnNumber):
        # Key Insight:
        # This is a base-26 system
        # A to Z -> 26

        results = []

        # We shift columnNumber by 1
        columnNumber -= 1

        while True:
            r = columnNumber % 26
            ch = chr(ord('A') + r)

            results.append(ch)

            columnNumber //= 26
            columnNumber -= 1

            if columnNumber < 0:
                break

        return "".join(results[::-1])
```

```python
solution = Solution()
```

```python
columnNumber = 1

print(solution.convertToTitle(columnNumber))
```
