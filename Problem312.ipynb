```python
class Solution:
    def maxCoins(self, nums):
        # Use Interval DP (Matrix Chain Multiplication Style)

        # Idea:
        # Instead of asking:
        # - Which baloon do I burst first?
        # we ask:
        # - Which balloon do I burst last?

        # Note: we burst last ballon in the interval, not the entire array.
        #
        # e.g. nums = [3,1,5]
        # after padding: a = [1,3,1,5,1]
        # The DP is not asking:
        # - What if the middle 1 is the last balloon in the entire game?
        # Instead it's asking:
        # - What if the middle 1 is the last balloon burst between 3 and 5?

        # For the subproblem of this problem promises:
        # - If we're solving the interval between two boundary balloons, those
        # boundaries stay alive until the interval is completely solved.


        # Pad the array with 1s on both ends
        a = [1] + nums + [1]
        n = len(a)

        # dp[i][j] will store the max coins from bursting baloons strictly between i and j
        # Note: not including i, not including j.
        # Since we define dp[i][j] as bursting balloons strictly i and j, when you burst
        # balloons within the interval, i and j are guaranteed to be alive.
        # Everything between i and j is being burst but i and j stay alive.
        # So at the final moment, the situation must look like:
        # [i ... k ... j]
        # That's why the last burst gives:
        # a[i] * a[k] * a[j]
        # And for everything else, before we burst k last,
        # everything between i and k, and k and j must already be fully burst.
        # Those are:
        # - dp[i][k]: left side of k that everything between i and k already be fully burst
        # - dp[k][j]: right side of k that everything between k and j already be fully burst
        # We add dp[i][k] and dp[k][j] because these are independent parts of history
        # before the final explosion at k.
        # Thus, the total coins if k is last:
        # - dp[i][j] = dp[i][k] + dp[k][j] + a[i] * a[k] * a[j]

        # By picking k, we force k to be the last balloon, everything else
        # splits into two independent subproblems.
        # Fixing k as the last burst cuts the interval into two disjoint
        # subproblems that no longer interact.
        dp = [[0] * n for _ in range(n)]

        # length is the length of the window of balloons we are bursting
        # It grows from a window of 1 baloon up to the full length of nums
        for length in range(1, len(nums) + 1):
            # i is the left boundary (exclusive)
            for i in range(0, n - length - 1):
                # j is the right boundary (exclusive)
                j = i + length + 1

                # Find the optimal last balloon 'k' to burst in the range (i, j)
                for k in range(i + 1, j):
                    coins = dp[i][k] + dp[k][j] + (a[i] * a[k] * a[j])

                    if coins > dp[i][j]:
                        dp[i][j] = coins

        # The answer for the full array is from index 0 to n - 1
        return dp [0][n - 1]
```

```python
solution = Solution()
```

```python
nums = [3,1,5,8]

print(solution.maxCoins(nums))
```

```python
nums = [1,5]

print(solution.maxCoins(nums))
```
