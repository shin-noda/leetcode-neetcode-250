```python
class Solution:
    def islandPerimeter(self, grid):
        peri = 0

        for i in range(len(grid)):
            for j in range(len(grid[i])):
                if grid[i][j] == 1:
                    peri += 4

                    # Check left
                    if j + 1 < len(grid[i]) and grid[i][j + 1] == 1:
                        peri -= 2

                    # Check bottom
                    if i + 1 < len(grid) and grid[i + 1][j] == 1:
                        peri -= 2

        return peri
```
