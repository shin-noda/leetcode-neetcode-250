```python
import heapq

class Solution:
    def lastStoneWeight(self, stones):
        # We negate all elements in the stones first and heapify them
        # so that it's O(N) instead of O(NlogN)
        for i in range(len(stones)):
            # Note: it's a max heap, so multiply by -1
            stones[i] = -stones[i]

        max_heap = stones
        heapq.heapify(max_heap)

        while len(max_heap) >= 2:
            y = -heapq.heappop(max_heap)
            x = -heapq.heappop(max_heap)

            if x != y:
                heapq.heappush(max_heap, -(y - x))

        if max_heap:
            return -max_heap[0]

        return 0
```
