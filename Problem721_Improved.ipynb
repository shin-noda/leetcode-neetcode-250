```python
class UnionFind:
    def __init__(self):
        self.parent = {}


    def find(self, i):
        if self.parent[i] != i:
            # Path compression
            self.parent[i] = self.find(self.parent[i])

        return self.parent[i]


    def union(self, i, j):
        root_i = self.find(i)
        root_j = self.find(j)

        if root_i != root_j:
            # root_j is the parent here
            self.parent[root_i] = root_j


class Solution:
    def accountsMerge(self, accounts):
        uf = UnionFind()
        email_to_name = {}

        # Initialize parents and map emails to names
        for account in accounts:
            name = account[0]
            for email in account[1:]:
                if email not in uf.parent:
                    # If we've never seen this email, it starts alone.
                    # a -> a, b -> b, c -> c, etc.
                    uf.parent[email] = email

                email_to_name[email] = name

                # Union the first email of this account with all other emails in it
                # e.g. a -> a, a -> b, a -> c
                uf.union(account[1], email)

        # Group emails by their root representative
        merged_accounts = {}
        for email in uf.parent:
            root = uf.find(email)
            if root not in merged_accounts:
                merged_accounts[root] = []

            merged_accounts[root].append(email)

        # Format the output (Sort emails, prepend name)
        result = []
        for root, emails in merged_accounts.items():
            result.append([email_to_name[root]] + sorted(emails))

        return result
```
