```python
class Solution:
    def longestIncreasingPath(self, matrix):
        if not matrix or not matrix[0]:
            return 0

        rows, cols = len(matrix), len(matrix[0])

        # dp[r][c] will store the longest path starting at (r, c)
        dp = [[0] * cols for _ in range(rows)]


        def dfs(r, c):
            # If already computed, return the cached result immediately
            if dp[r][c] != 0:
                return dp[r][c]

            # A single cell has a path length of 1
            max_path = 1

            # Explore all 4 neighbours
            for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
                nr, nc = r + dr, c + dc

                # Check boundaries and ensure the next cell is strictly greater
                if (
                    0 <= nr < rows and
                    0 <= nc < cols and
                    matrix[nr][nc] > matrix[r][c]
                ):
                    max_path = max(max_path, 1 + dfs(nr, nc))

            # Cache the final maximum result for this cell
            dp[r][c] = max_path

            return dp[r][c]

        max_val = 0
        for r in range(rows):
            for c in range(cols):
                max_val = max(max_val, dfs(r, c))

        return max_val
```

```python
solution = Solution()
```

```python
matrix = [[9,9,4],[6,6,8],[2,1,1]]

print(solution.longestIncreasingPath(matrix))
```
