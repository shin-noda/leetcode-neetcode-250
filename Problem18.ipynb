```python
class Solution:
    def fourSum(self, nums, target):
        # Sort the array
        nums.sort()

        n = len(nums)
        results = []

        # We fix the left mostpointer i.
        for i in range(n - 3):
            # Skip duplicate valus for i
            if i > 0 and nums[i] == nums[i - 1]:
                continue

            # We also fix the 2nd leftmost pointer j
            for j in range(i + 1, n - 2):
                # Skip duplicate values for j
                if j > i + 1 and nums[j] == nums[j - 1]:
                    continue

                # Use the Two pointer approach
                left = j + 1
                right = n - 1
                while left < right:
                    total = nums[i] + nums[j] + nums[left] + nums[right]

                    # Matched
                    if total == target:
                        results.append([nums[i], nums[j], nums[left], nums[right]])

                        # Skip duplicate values for Left
                        while left < right and nums[left] == nums[left + 1]:
                            left += 1

                        # Skip duplicate values for Right
                        while left < right and nums[right] == nums[right - 1]:
                            right -= 1

                        # Move on to the next loop
                        left += 1
                        right -= 1

                    # We need more value
                    elif total < target:
                        left += 1

                    # We overshot (total > target)
                    else:
                        right -= 1

        return results
```

```python
solution = Solution()
```

```python
nums = [1,0,-1,0,-2,2]
target = 0

print(solution.fourSum(nums, target))
```

```python
nums = [2,2,2,2,2]
target = 8

print(solution.fourSum(nums, target))
```
