```python
class Solution:
    def transpose(self, matrix):
        rows, cols = len(matrix), len(matrix[0])

        # We need to create a new matrix because it's not guaranteed that
        # 'matrix' is square
        transposed = [[0] * rows for _ in range(cols)]

        for r in range(rows):
            for c in range(cols):
                transposed[c][r] = matrix[r][c]

        return transposed
```
