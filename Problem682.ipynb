```python
class Solution:
    def calPoints(self, operations):
        scores = []

        for op in operations:
            # Note: .isdigit() does not work for negative numbers
            if op.isdigit() or op[0] == '-':
                scores.append(int(op))

            elif op == '+':
                prev1, prev2 = scores[-1], scores[-2]
                scores.append(prev1 + prev2)

            elif op == 'D':
                prev1 = scores[-1]
                scores.append(prev1 * 2)

            elif op == 'C':
                scores.pop()

        sum_val = 0
        if scores:
            sum_val = sum(scores)

        return sum_val
```

```python
solution = Solution()
```

```python
ops = ["5","2","C","D","+"]

print(solution.calPoints(ops))
```

```python
ops = ["5","-2","4","C","D","9","+","+"]

print(solution.calPoints(ops))
```

```python
ops = ["1","C"]

print(solution.calPoints(ops))
```
