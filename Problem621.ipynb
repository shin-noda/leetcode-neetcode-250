```python
class Solution:
    def leastInterval(self, tasks, n):
        # Idea
        # For a brute force approach, we can do something like this:
        # tasks = ["A","A","A","B","B","B"], n = 2
        # First, sort the elements that has the largest frequency.
        # In this example above, it's already sorted.
        # Next, we place first element like this:
        # A idle idle A idle idle A idle idle
        # Then, place the second element
        # A B idle A B idle A B idle
        # Finally, trim the idels at the end
        # A B idle A B idle A B
        # The answer is 8.
        #
        # We can implement this idea by
        # -using mashmap that has a letter as a key and occurences as a value
        # -use tuple packing and create a list consists of something like [(3, A), (54, B), ...]
        # -use a max heap and take it, decrement it, and push it back.
        # -when the occurence is zero, pop it.

        char_map = {}
        intervals = 0

        # O(N) where N is len(tasks).
        for ch in tasks:
            char_map[ch] = char_map.get(ch, 0) + 1

        # O(26) -> O(1)
        occurences = []
        for key in char_map:
            val = char_map[key]
            occurences.append((val, key))

        # Sort: O(NlogN) = O(26log26) -> O(1)
        # Non-decreasing (ascending) order
        oc = sorted(occurences, key=lambda x: x[0])

        # Get the max value from the end
        max_oc = oc.pop()[0]

        # Minimum intervals
        intervals = (n + 1) * (max_oc - 1) + 1

        # O(26) -> O(1)
        while oc and oc[-1][0] == max_oc:
            oc.pop()
            intervals += 1

        return max(intervals, len(tasks))
```

```python
solution = Solution()
```

```python
tasks = ["A","A","A","B","B","B"]
n = 2

print(solution.leastInterval(tasks, n))
```
