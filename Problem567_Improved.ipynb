```python
class Solution:
    def checkInclusion(self, s1, s2):
        if len(s1) > len(s2):
            return False

        # Use arrays of size 26 for lowercase English letters
        s1_counts = [0] * 26
        s2_counts = [0] * 26

        for i in range(len(s1)):
            s1_counts[ord(s1[i]) - ord('a')] += 1
            s2_counts[ord(s2[i]) - ord('a')] += 1

        # Count how many characters initially match
        matches = 0
        for i in range(26):
            if s1_counts[i] == s2_counts[i]:
                matches += 1

        # Slide the window
        left = 0
        for right in range(len(s1), len(s2)):
            if matches == 26:
                return True

            # Process the character entering from the right
            index = ord(s2[right]) - ord('a')
            s2_counts[index] += 1

            # When we add a new character to our window, only
            # two things can happen:
            # 1: The success case: the new character makes the count in s2
            # exactly equal to the count in s1.
            # 2: The failure case: the count in s2 was already equal to s1, but
            # adding this new character makes it too high.
            # That's why it uses if and elif here.
            if s1_counts[index] == s2_counts[index]:
                matches += 1

            elif s1_counts[index] + 1 == s2_counts[index]:
                matches -=  1

            # Process the character leaving from the left
            index = ord(s2[left]) - ord('a')
            s2_counts[index] -= 1
            if s1_counts[index] == s2_counts[index]:
                matches += 1

            elif s1_counts[index] - 1 == s2_counts[index]:
                matches -= 1

            left += 1

        return matches == 26
```

```python
solution = Solution()
```
