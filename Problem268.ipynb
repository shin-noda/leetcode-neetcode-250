```python
class Solution:
    def missingNumber(self, nums):
        missing = 0
        for i in range(len(nums)):
            missing += (i + 1) - nums[i]

        return missing
```

```python
solution = Solution()
```

```python
nums = [3,0,1]

print(solution.missingNumber(nums))
```

```python
nums = [0,1]

print(solution.missingNumber(nums))
```

```python
nums = [9,6,4,2,3,5,7,0,1]

print(solution.missingNumber(nums))
```
