```python
class Solution:
    def findItinerary(self, tickets):
        # This is an Eulerian Path problem.
        # Use Hierholzer's Algorithm

        # Step 1: Build the graph using a standard dictionary
        adj = {}
        for src, dst in tickets:
            if src not in adj:
                adj[src] = []
            adj[src].append(dst)

        # Step 2: Sort in reverse lexical order
        # This allows us to use pop() O(1) to grab the smallest lexical airport
        for src in adj:
            adj[src].sort(reverse=True)

        itinerary = []

        # Step 3: Post-order-DFS
        def dfs(airport):
            # Note: the base case is implicitly handled by the while loop condition.

            # We use .get() to avoid KeyError if an airport has no outgoign flights
            while airport in adj and adj[airport]:
                next_dest = adj[airport].pop()
                dfs(next_dest)

            # Pop complete / dead end reached
            itinerary.append(airport)

        # Start the journey at JFK
        dfs("JFK")

        # Step 4: Reverse the result
        return itinerary[::-1]
```
