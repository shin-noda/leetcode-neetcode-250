```python
class MyQueue:

    def __init__(self):
        # Those lists only behave like stacks
        self.in_stack = []
        self.out_stack = []


    def push(self, x):
        self.in_stack.append(x)


    def pop(self):
        # Ensure out_stack is populated
        self.peek()
        return self.out_stack.pop()


    def peek(self):
        # If the output gate is empty, move everything over
        if not self.out_stack:
            while self.in_stack:
                self.out_stack.append(self.in_stack.pop())

        return self.out_stack[-1]


    def empty(self):
        return not self.in_stack and not self.out_stack
```
