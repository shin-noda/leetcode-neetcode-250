```python
class Solution:
    def isMatch(self, s, p):
        n, m = len(s), len(p)

        # Create a grid initialized to False
        # Each cell answers:
        # If we start from position 'i' in 's' and 'j' in 'p', can
        # we match everything perfectly?
        dp = [[False] * (m + 1) for _ in range(n + 1)]

        # dp[n][m] represents empty string matching empty pattern (True)
        dp[n][m] = True

        # Loop backwards through 's' and 'p'
        for i in range(n, -1, -1):
            for j in range(m - 1, -1, -1):
                # Check if the current characters match
                first_match = (i < n) and (p[j] == s[i] or p[j] == '.')

                # Scenario 1: We encounter a Kleene Star
                if j + 1 < m and p[j + 1] == '*':
                    # Either
                    # - Skip the star or
                    # - Consume character if it matched
                    dp[i][j] = dp[i][j + 2] or (first_match and dp[i + 1][j])

                # Scenario 2: No star, move diagonally upwards
                else:
                    # Note: we move upwards
                    dp[i][j] = first_match and dp[i + 1][j + 1]

        # The answer to the full string and pattern is stored at the start
        return dp[0][0]
```

```python
solution = Solution()
```
