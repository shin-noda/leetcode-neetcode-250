```python
class Solution:
    def integerBreak(self, n):
        dp = [0] * (n + 1)

        # Note: 3 = 2 + 1, 2 * 1 = 2
        # dp[3] = 2

        # Note: 2 * 2 = 4
        # 3 * 1 = 3 is less
        # dp[4] = 4

        # 3 * 2 = 6
        # 4 * 1 = 4 is less
        # dp[5] = 6

        # 3 * 3 = 9
        # 4 * 2 = 8
        # dp[6] = 9

        # 4 * 3 = 12
        # 5 * 2 = 10
        # 6 * 1 = 6
        # dp[7] = 12

        # 2 * 3 * 3 = 18
        # 4 * 4 = 16
        # 5 * 3 = 15
        # dp[8] = 18

        if n == 2:
            return 1

        elif n == 3:
            return 2

        # Base cases
        dp[2] = 1
        dp[3] = 2
        dp[4] = 4

        for i in range(5, n + 1):
            for j in range(2, (n // 2) + 1):
                dp[i] = max(
                    dp[i],
                    dp[j] + dp[i - j],
                    j * (i - j),
                    j * dp[i - j],
                    dp[j] * (i - j)
                )

        return dp[n]
```

```python
solution = Solution()
```

```python
n = 2

print(solution.integerBreak(n))
```

```python
n = 10

print(solution.integerBreak(n))
```
