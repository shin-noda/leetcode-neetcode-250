```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None
```

```python
class Codec:
    def serialize(self, root):
        """Encodes a tree to a single string.

        :type root: TreeNode
        :rtype: str
        """

        results = []

        def dfs(node):
            if not node:
                results.append("None")
                return

            results.append(str(node.val))
            dfs(node.left)
            dfs(node.right)

        dfs(root)

        return ",".join(results)


    def deserialize(self, data):
        """Decodes your encoded data to tree.

        :type data: str
        :rtype: TreeNode
        """

        if not data:
            return None

        nodes = data.split(",")

        # This is a Generator Function
        def node_provider():
            for node in nodes:
                yield node

        # Instantiate the generator
        token_stream = node_provider()

        def dfs():
            # Get the next token from the generator
            val = next(token_stream)

            if val == "None":
                return None

            node = TreeNode(int(val))
            node.left = dfs()
            node.right = dfs()

            return node

        return dfs()
```
