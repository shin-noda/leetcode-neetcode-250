```python
class Solution:
    def subsetsWithDup(self, nums):
        results = []
        solutions = []

        # Group duplicates together
        nums.sort()

        def backtrack(i):
            # Base Case
            if i == len(nums):
                results.append(solutions[:])
                return

            # Choice 1: Take it
            solutions.append(nums[i])
            backtrack(i + 1)
            solutions.pop()

            # Choice 2: Leave it
            # Fast-forward past all identical items on the belt
            while i + 1 < len(nums) and nums[i] == nums[i + 1]:
                i += 1

            # Now 'i' is sitting at the last duplicate, so
            # i + 1 will point to a brand new element
            backtrack(i + 1)

        backtrack(0)

        return results
```
