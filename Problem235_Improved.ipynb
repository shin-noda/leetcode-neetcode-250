```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None
```

```python
class Solution:
    def lowestCommonAncestor(self, root, p, q):
        curr = root

        while curr:
            # Case 1: Both p.val and q.val are greater than curr.val -> move right
            if p.val > curr.val and q.val > curr.val:
                curr = curr.right

            # Case 2: Both are smaller -> move left
            elif p.val < curr.val and q.val < curr.val:
                curr = curr.left

            # Case 3: We found the split or hit p/q directly
            else:
                return curr
```
